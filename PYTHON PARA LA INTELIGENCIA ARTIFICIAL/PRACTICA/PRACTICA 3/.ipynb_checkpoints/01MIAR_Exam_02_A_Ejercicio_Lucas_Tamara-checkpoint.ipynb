### 📝 Explicación detallada del punto 3.2

**Qué pide exactamente el enunciado:**
Rellenar los `NaN` de la columna `age` con la **mediana de la edad calculada por clase (`pclass`)**, no con una única mediana global. Es decir: un pasajero de `pclass == 1` con edad desconocida debe rellenarse con la mediana de edad de los pasajeros de 1ª clase; uno de `pclass == 3`, con la mediana de 3ª clase. Esto evita el sesgo de usar "la mediana de todo el barco" cuando la edad no se distribuye igual en cada clase.

⚠️ Revisa tu celda anterior: agrupaste por `survived` en vez de por `pclass`, y aplicaste el relleno sobre la columna `pclass` en vez de sobre `age`. Las columnas y la agrupación deben coincidir con lo que pide el enunciado.

**Cómo funciona el patrón `groupby` + `transform` + `fillna`:**

1. **`df_clean.groupby('pclass')`** — agrupa las filas del DataFrame en "sub-tablas" según su valor de `pclass` (una para clase 1, otra para clase 2, otra para clase 3).
2. **`['age']`** — dentro de cada grupo, te quedas solo con la columna `age`.
3. **`.transform('median')`** — calcula la mediana de `age` **dentro de cada grupo**, ignorando los `NaN` al calcularla, pero el resultado que devuelve tiene **el mismo tamaño e índice que el DataFrame original** (a diferencia de `.agg('median')`, que colapsaría cada grupo en un solo valor). Es decir, repite el valor de la mediana de su grupo en cada fila correspondiente.
4. **`.fillna(...)`** — recorre `age` fila a fila: si el valor es `NaN`, lo sustituye por el valor (en esa misma posición) de la serie generada en el paso 3; si `age` ya tiene un valor, no lo toca.

**Ejemplo con un DataFrame de juguete:**

```python
df_juguete = pd.DataFrame({
    'pclass': [1, 1, 3, 3, 3],
    'age':    [40, np.nan, 20, np.nan, 25]
})

df_juguete.groupby('pclass')['age'].transform('median')
# 0    40.0   <- mediana de clase 1 (40 y NaN -> mediana 40)
# 1    40.0
# 2    22.5   <- mediana de clase 3 (20, NaN, 25 -> mediana 22.5)
# 3    22.5
# 4    22.5
```

Al aplicar `.fillna()` con esa serie, solo se sustituyen las posiciones donde `age` era `NaN`; el resto queda igual.

**Verificación:** después de aplicarlo sobre la columna correcta, comprueba que ya no queden nulos:
```python
df_clean['age'].isnull().sum()  # debería dar 0
```
