## 5. Visualización Preliminar

Sin tocar nada todavía. Quiero ver la distribución de tipos, el crecimiento del catálogo a lo largo del tiempo y cómo se distribuyen los ratings — para entender la forma general del dataset antes de limpiar.

```python
# ─── Preparar datos para gráficos preliminares ────────────────────────────────
df_plot = df_raw.copy()
df_plot['date_added'] = pd.to_datetime(df_plot['date_added'].str.strip(), errors='coerce', format='mixed')
df_plot['add_year'] = df_plot['date_added'].dt.year

# ─── Figura 1: Tipo de contenido + Crecimiento por año ───────────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico de dona: Movies vs TV Shows
tipo_conteo  = df_plot['type'].value_counts()
colores_tipo = ['#E50914', '#221F1F']
wedges, texts, autotexts = axes[0].pie(
    tipo_conteo.values,
    labels=tipo_conteo.index,
    colors=colores_tipo,
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.82,
    wedgeprops=dict(width=0.55, edgecolor='white', linewidth=2),
    textprops={'fontsize': 13}
)
for at in autotexts:
    at.set_fontsize(12)
    at.set_fontweight('bold')
axes[0].set_title('Distribución del Catálogo\nMovies vs TV Shows', fontsize=14, fontweight='bold', pad=15)

# Gráfico de líneas: contenido añadido por año y tipo
contenido_por_año = df_plot.groupby(['add_year', 'type']).size().unstack(fill_value=0)
contenido_por_año = contenido_por_año[contenido_por_año.index.notna()]
contenido_por_año = contenido_por_año[
    (contenido_por_año.index >= 2008) & (contenido_por_año.index <= 2021)
]

for tipo, color in zip(['Movie', 'TV Show'], ['#E50914', '#4A90E2']):
    if tipo in contenido_por_año.columns:
        axes[1].plot(contenido_por_año.index, contenido_por_año[tipo],
                     marker='o', color=color, linewidth=2.2, markersize=6, label=tipo)
        axes[1].fill_between(contenido_por_año.index, contenido_por_año[tipo],
                             alpha=0.12, color=color)

axes[1].set_title('Crecimiento del Catálogo Netflix\npor Año de Incorporación', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Año', fontsize=12)
axes[1].set_ylabel('Títulos añadidos', fontsize=12)
axes[1].legend(fontsize=12)
axes[1].xaxis.set_major_locator(mticker.MultipleLocator(2))
axes[1].grid(True, alpha=0.4)

plt.tight_layout()
plt.savefig('viz_preliminar_1.png', dpi=150, bbox_inches='tight')
plt.show()
print('Figura 1: Distribución Movies/TV Shows y evolución del catálogo por año.')
```

```python
# ─── Figura 2: Ratings + Top géneros (vista preliminar) ──────────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Distribución de ratings
rating_conteo  = df_raw['rating'].value_counts().dropna().head(12)
colores_rating = sns.color_palette('husl', len(rating_conteo))
bars = axes[0].barh(rating_conteo.index[::-1], rating_conteo.values[::-1],
                    color=colores_rating[::-1], edgecolor='white', linewidth=0.8)
for bar, val in zip(bars, rating_conteo.values[::-1]):
    axes[0].text(val + 15, bar.get_y() + bar.get_height() / 2,
                 f'{val:,}', va='center', fontsize=10, fontweight='bold')
axes[0].set_title('Distribución de Ratings\n(clasificación por edad)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Número de títulos', fontsize=12)
axes[0].set_xlim(0, rating_conteo.max() * 1.15)
axes[0].grid(True, axis='x', alpha=0.4)

# Top 10 géneros (exploración rápida)
generos_raw = (
    df_raw['listed_in'].str.split(', ').explode().str.strip().value_counts().head(10)
)
axes[1].barh(generos_raw.index[::-1], generos_raw.values[::-1],
             color=sns.color_palette('husl', 10)[::-1], edgecolor='white')
for i, val in enumerate(generos_raw.values[::-1]):
    axes[1].text(val + 20, i, f'{val:,}', va='center', fontsize=10, fontweight='bold')
axes[1].set_title('Top 10 Géneros Más Frecuentes\n(preliminar)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Apariciones', fontsize=12)
axes[1].set_xlim(0, generos_raw.max() * 1.12)
axes[1].grid(True, axis='x', alpha=0.4)

plt.tight_layout()
plt.savefig('viz_preliminar_2.png', dpi=150, bbox_inches='tight')
plt.show()
print('Figura 2: Distribución de ratings y top géneros sin preprocesamiento.')
```

## 7. Procedimiento

### 7.1 Limpieza de Datos

Trabajo sobre una copia del dataset original. Las decisiones de imputación dependen del contexto de cada columna — no hay una receta única para todos los nulos.

```python
# ─── Copia de trabajo ─────────────────────────────────────────────────────────
df = df_raw.copy()

# ─── 1. Imputar nulos en columnas de metadatos con 'Unknown' ──────────────────
# director, cast y country son optativos en la ficha del título —
# sin ellos el registro sigue siendo válido para análisis de tipo, género y fecha
for col in ['director', 'cast', 'country']:
    df[col] = df[col].fillna('Unknown')

# ─── 2. Eliminar filas donde date_added, rating o duration son nulos ──────────
# Son menos del 1% del dataset — mejor eliminar que imputar con valores arbitrarios
n_antes = len(df)
df = df.dropna(subset=['date_added', 'rating', 'duration'])
print(f'Filas eliminadas (date_added / rating / duration nulos): {n_antes - len(df)}')

# ─── 3. Convertir date_added de string a datetime ─────────────────────────────
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), format='mixed')

# ─── 4. Estandarizar release_year como entero ─────────────────────────────────
df['release_year'] = df['release_year'].astype(int)

# ─── Verificación final ────────────────────────────────────────────────────────
print(f'\nDataset limpio : {df.shape[0]:,} filas × {df.shape[1]} columnas')
print(f'Valores nulos  : {df.isnull().sum().sum()}')
df[['title', 'type', 'date_added', 'rating', 'duration']].head(6)
```

### 7.2 Feature Engineering

La columna `duration` mezcla formatos completamente distintos según el tipo: `"90 min"` para películas y `"2 Seasons"` para series. Hay que separar el número de la unidad para poder operar sobre duraciones de forma numérica.

Además, descompongo `date_added` en año y mes para analizar estacionalidad en las incorporaciones al catálogo.

```python
# ─── Extraer duración numérica y unidad desde 'duration' ──────────────────────
# Regex: captura el número y la unidad textual que le sigue
df['duration_val']  = df['duration'].str.extract(r'(\d+)').astype(float)
df['duration_unit'] = df['duration'].str.extract(r'\d+\s+(.+)')

# Normalizar: "Season" → "Seasons" (algunos registros usan singular)
df['duration_unit'] = df['duration_unit'].str.strip().str.replace(r'^Season$', 'Seasons', regex=True)

# ─── Extraer año y mes desde date_added ───────────────────────────────────────
df['add_year']  = df['date_added'].dt.year
df['add_month'] = df['date_added'].dt.month

# ─── Resumen de nuevas columnas ───────────────────────────────────────────────
print('── Nuevas características creadas ──')
print(f"  duration_val  : min={df['duration_val'].min():.0f}, "
      f"max={df['duration_val'].max():.0f}, "
      f"media={df['duration_val'].mean():.1f}")
print(f"  duration_unit : {df['duration_unit'].value_counts().to_dict()}")
print(f"  add_year      : {int(df['add_year'].min())} → {int(df['add_year'].max())}")
print(f"  add_month     : 1 → 12")

df[['title', 'type', 'duration', 'duration_val', 'duration_unit', 'add_year', 'add_month']].head(8)
```

### 7.3 Análisis de Géneros

La columna `listed_in` contiene géneros separados por coma — un título puede tener entre 1 y 3 géneros asignados. Para contar géneros reales necesito explotar la columna: convertir cada lista en filas individuales antes de agregar.

```python
# ─── Explotar géneros (un género por fila) ────────────────────────────────────
genres_series = df['listed_in'].str.split(', ').explode().str.strip()
top_genres    = genres_series.value_counts().head(15)

# ─── Géneros por tipo de contenido ────────────────────────────────────────────
df_genres = df[['type', 'listed_in']].copy()
df_genres = df_genres.assign(
    genre=df_genres['listed_in'].str.split(', ')
).explode('genre')
df_genres['genre'] = df_genres['genre'].str.strip()

genre_por_tipo = (
    df_genres.groupby(['genre', 'type'])
    .size()
    .unstack(fill_value=0)
)
genre_por_tipo['Total'] = genre_por_tipo.sum(axis=1)
genre_por_tipo = genre_por_tipo.sort_values('Total', ascending=False).head(15)

print('── Top 15 géneros por tipo de contenido ──')
print(genre_por_tipo.to_string())
```

### 7.4 Análisis por País

`country` también puede contener múltiples valores separados por coma — coproduciones internacionales. Aplico la misma lógica de explosión para contar la contribución real de cada país al catálogo.

```python
# ─── Explotar países (excluir 'Unknown') ─────────────────────────────────────
df_paises = df[df['country'] != 'Unknown'][['country', 'type']].copy()
df_paises = df_paises.assign(
    pais=df_paises['country'].str.split(', ')
).explode('pais')
df_paises['pais'] = df_paises['pais'].str.strip()

top_paises = df_paises['pais'].value_counts().head(10)

# ─── Top 10 países por tipo ────────────────────────────────────────────────────
pais_tipo = (
    df_paises.groupby(['pais', 'type'])
    .size()
    .unstack(fill_value=0)
)
pais_tipo['Total'] = pais_tipo.sum(axis=1)
pais_tipo = pais_tipo.sort_values('Total', ascending=False).head(10)

print('── Top 10 países productores de contenido Netflix ──')
print(pais_tipo.to_string())
```

### 7.5 Duración y Distribución de Ratings

Separo el análisis de duración por tipo: para películas me interesa la distribución de minutos, para series el número de temporadas. Los ratings los cruzo con el tipo para detectar diferencias editoriales entre ambos formatos.

```python
# ─── Estadísticas de duración por tipo ────────────────────────────────────────
print('── Duración de Películas (minutos) ──')
movies_dur = df[df['type'] == 'Movie']['duration_val']
print(movies_dur.describe().round(1).to_string())

print('\n── Número de Temporadas — TV Shows ──')
tv_dur = df[df['type'] == 'TV Show']['duration_val']
print(tv_dur.describe().round(1).to_string())

# ─── Ratings cruzados por tipo ────────────────────────────────────────────────
print('\n── Distribución de Ratings por Tipo de Contenido ──')
rating_pivot = df.groupby(['rating', 'type']).size().unstack(fill_value=0)
rating_pivot['Total'] = rating_pivot.sum(axis=1)
print(rating_pivot.sort_values('Total', ascending=False).to_string())
```

## 8. Visualización Final

Cuatro figuras que sintetizan los hallazgos del análisis: distribución geográfica, géneros por tipo, duración y clasificación de edad, y estacionalidad del catálogo. Cada una responde a una dimensión distinta del contenido Netflix.

```python
# ─── Figura 3: Top países + Top géneros ──────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# Top 10 países
colores_p = sns.color_palette('husl', 10)
bars_p = axes[0].barh(top_paises.index[::-1], top_paises.values[::-1],
                      color=colores_p[::-1], edgecolor='white', linewidth=0.8)
for bar, val in zip(bars_p, top_paises.values[::-1]):
    axes[0].text(val + 20, bar.get_y() + bar.get_height() / 2,
                 f'{val:,}', va='center', fontsize=10, fontweight='bold')
axes[0].set_title('Top 10 Países Productores de Contenido Netflix', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Número de títulos', fontsize=11)
axes[0].set_xlim(0, top_paises.max() * 1.15)
axes[0].grid(True, axis='x', alpha=0.4)

# Top 15 géneros
colores_g = sns.color_palette('husl', 15)
bars_g = axes[1].barh(top_genres.index[::-1], top_genres.values[::-1],
                      color=colores_g[::-1], edgecolor='white', linewidth=0.8)
for bar, val in zip(bars_g, top_genres.values[::-1]):
    axes[1].text(val + 15, bar.get_y() + bar.get_height() / 2,
                 f'{val:,}', va='center', fontsize=9, fontweight='bold')
axes[1].set_title('Top 15 Géneros del Catálogo Netflix', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Apariciones', fontsize=11)
axes[1].set_xlim(0, top_genres.max() * 1.12)
axes[1].grid(True, axis='x', alpha=0.4)

plt.tight_layout()
plt.savefig('viz_final_1_paises_generos.png', dpi=150, bbox_inches='tight')
plt.show()
print('Figura 3: Top 10 países productores y top 15 géneros del catálogo Netflix.')
```

**Hallazgos principales:**

- **Estados Unidos domina** la producción con más del doble que el segundo país (India). Pero el contenido internacional crece — "International Movies" es el género más frecuente, lo que refleja una estrategia deliberada de expansión global del catálogo.
- **Las películas son Drama e Internacional** como géneros primarios; las series apuntan a Crime TV Shows y Kids' TV — audiencias y formatos distintos con estrategias editoriales diferenciadas.
- **La duración típica de una película en Netflix es ~99 minutos** — ligeramente por encima del estándar de 90 min, con cola derecha de documentales y películas de acción largas.
- **`TV-MA` es el rating dominante** en ambos formatos — Netflix apuesta por contenido adulto maduro como producto estrella, con muy poco peso en clasificaciones para niños.
- **La estacionalidad existe**: julio, octubre y enero concentran más adiciones al catálogo — alineado con temporadas de consumo y ciclos de anuncio de contenido nuevo.
- **2019 fue el pico de incorporaciones** — antes del efecto COVID que ralentizó las producciones en 2020.

## 9. Reflexión

Lo que más me llamó la atención de este dataset no fue el contenido en sí, sino los problemas estructurales que generan las columnas multivalor. `listed_in` y `country` almacenan listas como strings — un diseño que funciona para presentar información en una tabla, pero que hace imposible agregar directamente. El proceso de explotar esas columnas con `str.split().explode()` es exactamente el tipo de transformación que en un pipeline de producción tendría que estar automatizado para que el análisis pueda escalar a cientos de miles de registros.

El manejo de nulos requirió criterio, no solo pandas. Para `director` la imputación con `'Unknown'` es correcta — no saber quién dirigió una película no invalida el registro para análisis de tipo, género o fecha. Para `date_added` y `rating` con menos del 1% de nulos, eliminar las filas fue más limpio que imputar con valores que podrían distorsionar los gráficos de distribución temporal.

La extracción de duración numérica con regex fue directa, pero evidenció un problema de diseño original: mezclar películas y series en la misma columna `duration` con unidades distintas. En un esquema relacional bien diseñado serían tablas separadas. Ese tipo de decisiones tiene un costo directo en el análisis posterior — y es un error que se comete más de lo que parece en datasets reales.

El catálogo de Netflix resultó ser un espejo de su estrategia de negocio: contenido adulto primario, foco en producción estadounidense con expansión internacional controlada, pico de adiciones en 2019 antes de que COVID frenara las producciones. Los datos no mienten — pero hay que saber limpiarlos, transformarlos y leerlos.
