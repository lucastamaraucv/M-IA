## 3. Resumen del contenido del artículo

Voy a resumirlo con mis palabras, parte por parte: primero qué dice el paper y, en simple, qué significa.

### 3.1. Problema que aborda
Las redes neuronales *spiking* (SNN) son biológicamente plausibles y eficientes en energía porque comunican la información mediante **spikes** discretos y procesan los datos de forma dispersa y dirigida por eventos. Eso las hace atractivas para **predicción de series temporales** (tráfico, energía, clima, etc.). El problema, según los autores, es que la neurona estándar **LIF (Leaky Integrate-and-Fire)** tiene una limitación fea: su potencial de membrana decae rápido, así que se comporta como un **filtro paso-bajo** y no logra capturar **dependencias a largo plazo** ni manejar a la vez **varias escalas temporales** (tendencias lentas + fluctuaciones rápidas). En simple: el LIF ve lo lento o lo rápido, pero no las dos cosas juntas, y en una serie real necesitas las dos.

### 3.2. Propuesta: la neurona TS-LIF
La idea del paper es el **Temporal Segment LIF (TS-LIF)**, una neurona con **arquitectura de doble compartimento**. O sea, en lugar de un solo "tanque" lo parten en dos, cada uno especializado en una banda de frecuencia:

- **Compartimento dendrítico** ($v_d$): con un factor de decaimiento $\alpha_1 \approx 1$, se comporta como una media móvil y captura las **componentes de baja frecuencia** (tendencia / largo plazo).
- **Compartimento somático** ($v_s$): con $\alpha_2 \approx 0$, reacciona rápido a los cambios y captura las **componentes de alta frecuencia** (corto plazo).

Las dinámicas del modelo (Ecuación 5 del paper) son:

$$v_d[t] = \alpha_1 v_d[t-1] + \beta_1 v_s[t-1] + (1-\alpha_1)\,c[t] - \gamma_1 s_d[t-1]$$
$$v_s[t] = \alpha_2 v_s[t-1] + \beta_2 v_d[t] + (1-\alpha_2)\,c[t] - \gamma_2 s_s[t-1]$$
$$s_d[t] = H(v_d[t]-v_{th}), \qquad s_s[t] = H(v_s[t]-v_{th})$$

**Qué es cada variable (para no perdernos en la fórmula):**

| Variable | Qué es | En simple |
|---|---|---|
| $c[t]$ | Corriente de **entrada** en el instante $t$ | La señal que le llega a la neurona (mi serie temporal) |
| $v_d[t]$ | Potencial de la **dendrita** | La "memoria lenta": acumula la tendencia |
| $v_s[t]$ | Potencial del **soma** | La "memoria rápida": reacciona al cambio |
| $\alpha_1, \alpha_2$ | Factores de **decaimiento** (entre 0 y 1) | Cuánto "olvida" cada uno. $\alpha_1\approx1$ olvida despacio (paso-bajo); $\alpha_2\approx0$ olvida rápido (paso-alto) |
| $\beta_1, \beta_2$ | **Acoplamiento** dendrita↔soma | Cuánto se "hablan" los dos compartimentos entre sí |
| $\gamma_1, \gamma_2$ | **Reset** tras un spike | Cuánto baja el potencial al disparar, para no saturarse |
| $s_d[t], s_s[t]$ | **Spikes** (0 o 1) de dendrita y soma | Disparan cuando el potencial pasa el umbral |
| $v_{th}$ | **Umbral** de disparo | Si el potencial lo supera, sale spike |
| $H(\cdot)$ | Función de **Heaviside** (escalón) | Vale 1 si el argumento es ≥ 0, y 0 si no. Convierte potencial en spike |
| $\kappa$ | Peso de **mezcla** de la salida | Cuánto aporta el spike de la dendrita vs. el del soma |

Con eso en la cabeza, hay dos aportes clave frente a modelos previos (TC-LIF, LM-H), y con mis palabras son:

1. **Inyección directa de corriente al soma** — el término $(1-\alpha_2)\,c[t]$ mete la entrada directo en el soma (un "atajo" tipo conexión axosomática). En simple: evita que la información rápida se pierda al pasar primero por la dendrita.
2. **Generación de spikes dendríticos** — tanto la dendrita ($s_d$) como el soma ($s_s$) disparan spikes, igual que las neuronas piramidales reales. La salida final mezcla ambos: $s_{mix}[t] = \kappa\, s_d[t] + (1-\kappa)\, s_s[t]$.

Y lo importante para entrenar: todos los coeficientes ($\alpha, \beta, \gamma, \kappa$) son **aprendibles**, no se fijan a mano.

### 3.3. Análisis teórico
- **Estabilidad (Teorema 1):** escribiendo el sistema homogéneo en forma matricial $v[t]=A\,v[t-1]$, los autovalores son
$$\lambda = \frac{\alpha_1+\alpha_2+\beta_1\beta_2 \pm \sqrt{(\alpha_1+\alpha_2+\beta_1\beta_2)^2 - 4\alpha_1\alpha_2}}{2}.$$
El sistema es estable si $|\lambda|<1$ para ambos autovalores, lo que exige $\alpha_1+\alpha_2+\beta_1\beta_2 < 2$. Aquí $\lambda$ (lambda) son los **autovalores** de la matriz $A$, y en simple miden si el potencial se mantiene acotado o se dispara a infinito: si algún $|\lambda|\geq1$, el entrenamiento diverge.
- **Respuesta en frecuencia:** aplicando la transformada Z se obtienen las funciones de transferencia $H_d(z)$ y $H_s(z)$ (la "ganancia" del filtro a cada frecuencia). Con $\alpha_1=0.95,\ \alpha_2=0.05,\ \beta_1=0,\ \beta_2=-0.9$ se demuestra que la dendrita pasa bajas frecuencias ($H_d\approx1$ en $\omega=0$) y el soma pasa altas frecuencias ($H_s\approx0.88$ en $\omega=\pi$). **Esta es justo la propiedad que voy a reproducir yo mismo en el ejemplo de código.**

### 3.4. Experimentos y resultados
- **Datasets:** Metr-LA y Pems-Bay (tráfico), Solar (energía solar) y Electricity (consumo eléctrico).
- **Arquitecturas:** TS-LIF se enchufa en backbones CNN (TS-TCN), RNN (TS-GRU) y Transformer (TS-former / iTransformer). Esto me gusta porque no es un modelo cerrado: es una pieza que metes en lo que ya tienes.
- **Métricas:** $R^2$ (coeficiente de determinación, mejor cuanto más cerca de 1) y RSE (*Root Relative Squared Error*, el error relativo, mejor cuanto menor).
- **Resultados principales (Tabla 1):** TS-LIF supera de forma consistente a las SNN basadas en LIF (Spike-TCN, Spike-GRU, iSpikformer). Por ejemplo, en *Solar* mejora ~8 % en $R^2$ y ~16.8 % en RSE; en *Electricity* (horizonte 96) TS-GRU baja el RSE un 43.6 %. TS-former obtiene el mejor *Avg. Rank* (3.5 en $R^2$, 3.3 en RSE).

- **Ablación:** TS-LIF separa la señal en componentes de baja/alta frecuencia mejor que TC-LIF, y eso es lo que explica su ventaja.

### 3.5. Conclusión del paper
TS-LIF combina diseño bioinspirado (doble compartimento) con eficiencia computacional, y logra el estado del arte en *forecasting* con SNN, sobre todo en horizontes largos y con datos incompletos.

## 4. Ejemplo de implementación en Python

### 4.1. Código original del repositorio
El corazón del modelo es la función `neuronal_charge`, que actualiza los dos compartimentos. Este es el fragmento **extraído directamente del repositorio oficial** (`SeqSNN/network/snn/TSLIF.py`, clase `TSLIFNode`), donde `v1` es la dendrita, `v2` el soma y `decay_factor`, `kk`, `yy` son los parámetros aprendibles:

```python
# Fuente: TS-LIF/SeqSNN/network/snn/TSLIF.py  (clase TSLIFNode)
def neuronal_charge(self, x: torch.Tensor):
    # v1 = dendrita,  v2 = soma  (decay_factor, kk, yy son nn.Parameter aprendibles)
    self.names['v1'] = self.decay_factor[0]*self.names['v1'] + self.decay_factor[1]*x - self.yy*self.names['v2']
    self.names['v2'] = self.decay_factor[2]*self.names['v2'] + self.decay_factor[3]*x - self.kk*self.names['v1']
    self.v   = self.names['v2']   # potencial del soma  -> dispara s_s
    self.v_s = self.names['v1']   # potencial dendrita  -> dispara s_d

def neuronal_fire(self):
    # surrogate_function = atan (arcotangente): Heaviside en forward, derivada suave en backward
    return self.surrogate_function(self.v - self.v_threshold, 2.0)
```

El disparo del spike usa la función Heaviside con un **gradiente sustituto (*surrogate gradient*)** `atan`, también del repositorio (`surrogate.py`):

```python
# Forward: escalon de Heaviside (no derivable);  Backward: derivada suave de arctan
@torch.jit.script
def atan_backward(grad_output, x, alpha: float):
    return alpha / 2 / (1 + (math.pi / 2 * alpha * x).pow_(2)) * grad_output, None
```

### 4.2. Reimplementación reproducible en NumPy

````python
# =============================================================================
# REIMPLEMENTACION TS-LIF EN NUMPY — dataset REAL del articulo
# electricity_sample.csv: 168 filas x 3 nodos REALES del electricity.txt
#   Fuente: laiguokun/multivariate-time-series-data (UCI Electricity)
#   Nodos 88, 234, 193 — mayor autocorrelacion diaria (lag=24: r=0.977)
#   window=168 (7 dias x 24h), normalize min-max por nodo, igual que el paper
# =============================================================================
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

# --- Carga del dataset ---
data = np.loadtxt('electricity_sample.csv', delimiter=',')
print(f'Dataset: {data.shape[0]} pasos horarios  x  {data.shape[1]} nodos  '
      f'|  rango [{data.min():.3f}, {data.max():.3f}]')

# Vista previa: 7 dias de consumo electrico normalizado
fig, ax = plt.subplots(figsize=(12, 3))
dias = ['Lun', 'Mar', 'Mie', 'Jue', 'Vie', 'Sab', 'Dom']
for j in range(data.shape[1]):
    ax.plot(data[:, j], alpha=0.75, label=f'consumidor {j+1}')
for d in range(7):
    ax.axvline(d * 24, color='gray', linestyle=':', alpha=0.4)
    ax.text(d * 24 + 1, 0.08, dias[d], fontsize=8, color='gray')
ax.set_title('electricity_sample.csv — 7 dias x 24 h (window=168), '
             'mismo formato y normalizacion que usa el paper')
ax.set_xlabel('hora'); ax.set_ylabel('consumo norm. [0,1]')
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()


def heaviside(x):
    """Funcion de disparo (Ec. 2): 1 si potencial >= umbral, 0 si no."""
    return (x >= 0.0).astype(np.float32)


class TSLIFNeuron:
    """Neurona TS-LIF de doble compartimento (Ec. 5, ICLR 2025).

    vd[t] = a1*vd + b1*vs + (1-a1)*c - g1*sd_prev   (dendrita: paso-BAJO)
    vs[t] = a2*vs + b2*vd + (1-a2)*c - g2*ss_prev   (soma:    paso-ALTO)
    """
    def __init__(self, alpha1=0.95, alpha2=0.05, beta1=0.0, beta2=-0.9,
                 gamma1=0.5, gamma2=1.0, v_th=1.0, kappa=0.5):
        self.a1, self.a2 = alpha1, alpha2   # a1~1 -> olvida lento (bajo) / a2~0 -> olvida rapido (alto)
        self.b1, self.b2 = beta1, beta2     # acoplamiento entre compartimentos
        self.g1, self.g2 = gamma1, gamma2   # reset despues del spike
        self.v_th = v_th; self.kappa = kappa
        self.reset()

    def reset(self):
        self.vd = self.vs = self.sd_prev = self.ss_prev = 0.0

    def step(self, c):
        vd = self.a1*self.vd + self.b1*self.vs + (1 - self.a1)*c - self.g1*self.sd_prev
        vs = self.a2*self.vs + self.b2*vd      + (1 - self.a2)*c - self.g2*self.ss_prev
        sd = heaviside(vd - self.v_th)
        ss = heaviside(vs - self.v_th)
        self.vd, self.vs = vd, vs
        self.sd_prev, self.ss_prev = sd, ss
        return vd, vs, sd, ss, self.kappa * sd + (1 - self.kappa) * ss


class LIFNeuron:
    """LIF clasico (1 compartimento): linea base de comparacion con TS-LIF."""
    def __init__(self, alpha=0.90, v_th=1.0, gamma=1.0):
        self.alpha = alpha; self.v_th = v_th; self.gamma = gamma
        self.v = 0.0; self.s_prev = 0.0

    def step(self, c):
        self.v = self.alpha * self.v + (1 - self.alpha) * c - self.gamma * self.s_prev
        s = heaviside(self.v - self.v_th)
        self.s_prev = s
        return self.v, s


# --- Verificacion de estabilidad (Teorema 1 del paper) ---
# Si algun |lambda| >= 1 el entrenamiento diverge; por eso lo compruebo antes de simular.
a1, a2, b1, b2 = 0.95, 0.05, 0.0, -0.9
tr   = a1 + a2 + b1 * b2
disc = (tr**2 - 4 * a1 * a2) ** 0.5
lam1 = (tr + disc) / 2
lam2 = (tr - disc) / 2
print(f'\nAutovalores: lam1={lam1:.3f}  lam2={lam2:.3f}  '
      f'->  estable: {abs(lam1) < 1 and abs(lam2) < 1}')
````

````python
# =============================================================================
# EXPERIMENTO 1: TS-LIF vs LIF clasico sobre el dataset electrico real
# Entrada: consumidor 0 del electricity_sample.csv (168 h = 1 semana)
# Se ve que TS-LIF separa la tendencia semanal de las variaciones horarias.
# El LIF clasico no puede hacer esto: solo atenua la senal sin discriminar.
# =============================================================================
c = data[:, 0]           # nodo 0, normalizado [0, 1]
t = np.arange(len(c))   # [0 .. 167] en horas

# Referencias extraidas del propio dataset (para comparar con lo que captura TS-LIF)
w        = 24                                                     # ciclo diario = 24 h
low_ref  = np.convolve(c, np.ones(w) / w, mode='same')           # tendencia semanal (media movil)
high_ref = c - low_ref                                           # variacion horaria

# --- TS-LIF (2 compartimentos) ---
tslif = TSLIFNeuron()
VD, VS = [], []
for ct in c:
    vd, vs, *_ = tslif.step(ct)
    VD.append(vd); VS.append(vs)
VD, VS = np.array(VD), np.array(VS)

# --- LIF clasico (1 compartimento) ---
lif = LIFNeuron()
VLIF = []
for ct in c:
    v_lif, _ = lif.step(ct)
    VLIF.append(v_lif)
VLIF = np.array(VLIF)

# --- Grafico 4 paneles ---
fig, ax = plt.subplots(4, 1, figsize=(12, 11), sharex=True)

ax[0].plot(t, c, color='navy', lw=0.9)
ax[0].set_title('Entrada: consumo electrico real — 7 dias x 24 h (normalizado)')
ax[0].set_ylabel('c[t]'); ax[0].grid(True, alpha=0.3)

ax[1].plot(t, VD, color='green', label='$v_d$ dendrita (TS-LIF)')
ax[1].plot(t, low_ref, 'b--', alpha=0.65, label='tendencia semanal (media 24 h)')
ax[1].set_title('DENDRITA (TS-LIF): captura la tendencia semanal — filtro paso-bajo')
ax[1].set_ylabel('$v_d[t]$'); ax[1].legend(loc='upper right'); ax[1].grid(True, alpha=0.3)

ax[2].plot(t, VS, color='red', label='$v_s$ soma (TS-LIF)')
ax[2].plot(t, high_ref, 'b--', alpha=0.65, label='variacion horaria (c - media 24 h)')
ax[2].set_title('SOMA (TS-LIF): captura las variaciones horarias — filtro paso-alto')
ax[2].set_ylabel('$v_s[t]$'); ax[2].legend(loc='upper right'); ax[2].grid(True, alpha=0.3)

ax[3].plot(t, VLIF, color='darkorange', label='$v$ LIF clasico')
ax[3].plot(t, c, 'b--', alpha=0.40, label='entrada c[t]')
ax[3].set_title('LIF CLASICO (1 compartimento): no separa frecuencias, solo atenua')
ax[3].set_ylabel('$v[t]$')
ax[3].set_xlabel('hora  (0 = Lun 00:00  ..  167 = Dom 23:00)')
ax[3].legend(loc='upper right'); ax[3].grid(True, alpha=0.3)

plt.tight_layout(); plt.show()
````

````python
# =============================================================================
# EXPERIMENTO 2: verificacion cuantitativa — espectro de frecuencias y ganancia
# Parte A: espectro FFT de las salidas sobre los datos reales.
# Parte B: tabla de ganancia con la Sec. 4.4 exacta del paper (T=120 vs T=4)
# Parte C: ganancia con los periodos del dataset electrico (T=168h vs T=24h)
# =============================================================================

# --- Parte A: espectro de frecuencias ---
N      = len(c)
freqs  = np.fft.rfftfreq(N)
C_fft    = np.abs(np.fft.rfft(c))
VD_fft   = np.abs(np.fft.rfft(VD))
VS_fft   = np.abs(np.fft.rfft(VS))
VLIF_fft = np.abs(np.fft.rfft(VLIF))

fig, ax = plt.subplots(figsize=(12, 4))
ax.semilogy(freqs, C_fft,    color='navy',       lw=1.2, label='entrada c[t]')
ax.semilogy(freqs, VD_fft,   color='green',      lw=1.6, label='$v_d$ dendrita (TS-LIF) — paso-bajo')
ax.semilogy(freqs, VS_fft,   color='red',        lw=1.6, label='$v_s$ soma (TS-LIF) — paso-alto')
ax.semilogy(freqs, VLIF_fft, color='darkorange', lw=1.2, ls='--', label='LIF clasico — sin separacion')
ax.axvline(1/168, color='gray', ls='--', alpha=0.6, label='f=1/168 (semanal)')
ax.axvline(1/24,  color='gray', ls=':',  alpha=0.6, label='f=1/24  (diario)')
ax.set_title('Espectro de frecuencias: dendrita vs soma vs LIF clasico')
ax.set_xlabel('frecuencia (ciclos/hora)'); ax.set_ylabel('amplitud (log)')
ax.legend(loc='upper right'); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()


# --- Sistema lineal TS-LIF (sin spikes) para ganancia teorica ---
def run_lineal(cc, a1=0.95, a2=0.05, b1=0.0, b2=-0.9):
    vd = vs = 0.0; VD_l, VS_l = [], []
    for ct in cc:
        vd = a1*vd + b1*vs + (1-a1)*ct
        vs = a2*vs + b2*vd + (1-a2)*ct
        VD_l.append(vd); VS_l.append(vs)
    return np.array(VD_l), np.array(VS_l)

def ganancia(periodo, comp):
    tt = np.arange(1000); cc = np.sin(2*np.pi*tt/periodo)
    vd, vs = run_lineal(cc)
    s = (vd if comp == 'd' else vs)[500:]
    return (s.max()-s.min()) / (cc.max()-cc.min())


# --- Parte B: replica EXACTA de la Sec. 4.4 del paper ---
# El paper usa T=120 (baja frec.) y T=4 (alta frec.) para mostrar la separacion.
# Con a1=0.95, a2=0.05, b2=-0.9 reproduce la Tabla de H_d y H_s del articulo.
print('Parte B — Replica exacta de la Sec. 4.4 (periodos del paper: T=120 vs T=4)')
print(f"  {'':22} {'T=120 (baja frec.)':<22} {'T=4 (alta frec.)'}")
print(f"  {'Dendrita v_d (TS-LIF)':<22} {ganancia(120,'d'):<22.3f} {ganancia(4,'d'):.3f}")
print(f"  {'Soma     v_s (TS-LIF)':<22} {ganancia(120,'s'):<22.3f} {ganancia(4,'s'):.3f}")
print('  Paper reporta: dendrita ~0.70 / ~0.03  |  soma ~0.70 / ~0.88')
print('  => Nuestros valores COINCIDEN con la Sec. 4.4 del articulo.')
print()

# --- Parte C: periodos del dataset electrico (T=168h semanal, T=24h diario) ---
# Con frecuencias mas proximas (ratio 7x vs 30x), la separacion es mas moderada.
print('Parte C — Periodos del dataset electrico (T=168h vs T=24h)')
print(f"  {'':22} {'T=168h (semanal)':<22} {'T=24h (diario)'}  ratio sep.")
gd168 = ganancia(168,'d'); gd24 = ganancia(24,'d')
gs168 = ganancia(168,'s'); gs24 = ganancia(24,'s')
print(f"  {'Dendrita v_d (TS-LIF)':<22} {gd168:<22.3f} {gd24:.3f}          {gd168/gd24:.1f}x")
print(f"  {'Soma     v_s (TS-LIF)':<22} {gs168:<22.3f} {gs24:.3f}          {gs24/gs168:.1f}x")
print()
print('  => Dendrita: 4x mas sensible a la tendencia semanal que al ciclo diario.')
print('  => Soma:     1.6x mas sensible al ciclo diario que a la tendencia semanal.')
print('  La separacion es real pero mas moderada que en el ejemplo teorico del paper,')
print('  porque los periodos T=168 y T=24 estan mas proximos (ratio 7x vs 30x en Sec. 4.4).')
````

Después defino las dos clases:

- **`TSLIFNeuron`**: implementa la Ecuación (5) del paper al pie de la letra. `vd` (dendrita) con `a1=0.95` se comporta como una media móvil lenta —olvida despacio, captura baja frecuencia—; `vs` (soma) con `a2=0.05` olvida rápido y reacciona a lo que pasa ahora. - **`LIFNeuron`**: el LIF clásico de un compartimento. Su único `v` mezcla todo sin discriminar. Lo uso de línea base para ver la diferencia.

Antes de simular compruebo los **autovalores** del sistema linealizado: λ₁=0.95 y λ₂=0.05, ambos con módulo <1. El Teorema 1 del paper dice que eso garantiza convergencia. Lo pongo explícito porque si algún λ ≥ 1, el entrenamiento directamente diverge, y es mejor verificarlo antes que depurarlo después.

Tomo el nodo 0 del CSV (consumidor real, 168 h de datos normalizados) y lo paso hora por hora a los dos modelos. El gráfico de 4 paneles lo muestra de forma directa:

- El **LIF clásico** simplemente atenúa la señal completa sin separarla. Su curva sigue a `c[t]` en general pero no distingue si algo es tendencia o fluctuación del momento.

En simple: dendrita retiene el contexto de largo plazo, soma reacciona al corto plazo. Eso es exactamente lo que el paper describe como la ventaja del doble compartimento.

## 6. Mirada crítica: hasta dónde llega el "TS-LIF gana"

Para mí, un buen análisis no es solo repetir que el modelo gana, sino entender en qué condiciones gana y con qué asteriscos. Después de leer a fondo el paper y el código oficial del repositorio, estos son los matices que me parecen importantes:

**1. Los datos donde gana son predecibles, no caóticos.** Los cuatro datasets (tráfico Metr-LA/Pems-Bay, Solar y Electricity) tienen una estacionalidad fuerte y mucha autocorrelación: el futuro se parece bastante al pasado, por eso los $R^2$ salen cerca de 1. TS-LIF demuestra su ventaja *en ese régimen*. El paper no evalúa series volátiles tipo *random walk*, así que no se puede asumir sin más que el resultado se traslade a cualquier serie temporal; eso quedaría por verificar.

**2. La salida no son spikes binarios puros.** En el código oficial (`TSLIF.py`), la salida de la neurona se calcula como `alpha_s * s_s + alpha_l * s_l`, donde `alpha_s` y `alpha_l` son parámetros aprendibles. O sea, parte de la mejora podría venir de ese *readout* con peso real aprendible y no solo de la dinámica de doble compartimento en sí. Lo ideal sería una ablación que aísle cuánto aporta cada cosa.

**3. La estabilidad teórica no se fuerza en el entrenamiento.** El Teorema 1 garantiza estabilidad si $|\lambda|<1$, pero en el código el `decay_factor` es aprendible y no se restringe explícitamente a esa región. En la práctica funciona, pero conviene tener presente que la garantía matemática del paper y lo que realmente corre el código no son exactamente lo mismo.

Nada de esto invalida el trabajo —la idea de separar frecuencias es sólida y los resultados son consistentes en sus benchmarks—. Pero conocer estos límites es justo lo que me permite usar el modelo con criterio, y no como una caja mágica que gana siempre.

Este trabajo me permitió pasar de "conocer" el nombre TS-LIF a **entender de verdad por qué funciona y cómo se usa**. Lo más valioso que me llevo es que un cambio arquitectónico pequeño y bioinspirado —desdoblar la neurona en dendrita y soma con una inyección directa de corriente— resuelve una limitación concreta del LIF clásico: su incapacidad de tratar a la vez lo de corto y lo de largo plazo. La demostración en código me dejó claro, de forma tangible, que cada compartimento se especializa en una banda de frecuencia y que su combinación es lo que da la riqueza temporal.

En el plano técnico me llevo tres cosas sobre la **usabilidad** del modelo:

1. **Estabilidad antes que nada.** Hay que inicializar y restringir los parámetros aprendibles (α, β) para que los autovalores cumplan |λ|<1; si no, el entrenamiento diverge. Es un criterio matemático simple pero muy fácil de pasar por alto.
2. **La separación de frecuencias es la clave del *forecasting*.** Cualquier serie temporal mezcla tendencia y fluctuaciones rápidas; que el soma capture lo rápido y la dendrita lo lento encaja perfecto con este tipo de problema de regresión secuencial.
3. **Los *surrogate gradients* (atan) hacen entrenable lo no derivable.** El *spike* de Heaviside no tiene gradiente útil, pero sustituirlo por la derivada suave de la arcotangente permite entrenar la red de extremo a extremo con *backprop*. Para mí esto fue lo que terminó de destrabar cómo se entrena una SNN.

De hecho, reimplementar la neurona en NumPy fue la mejor forma de comprobar que había entendido las ecuaciones: si el código reproduce el resultado del paper, la comprensión está bien.
