## if / elif / else

```python
if condicion:
    # ejecuta si True
elif otra_condicion:
    # ejecuta si la primera es False y esta True
else:
    # ejecuta si todas son False
```

La **indentación** (4 espacios) define los bloques. No hay llaves `{}` como en otros lenguajes.

**Expresión ternaria:**
```python
resultado = valor_si_true if condicion else valor_si_false
```

````python
nota = 7.5

if nota >= 9:
    cal = "Sobresaliente"
elif nota >= 7:
    cal = "Notable"
elif nota >= 5:
    cal = "Aprobado"
else:
    cal = "Suspenso"

print(f"Nota {nota}: {cal}")

# Ternario
edad = 20
status = "mayor" if edad >= 18 else "menor"
print(f"Con {edad} años eres {status} de edad")
````

## match-case (Python 3.10+)

Compara un valor con patrones. Más potente que if/elif para múltiples casos.

```python
match variable:
    case patron1:
        ...
    case patron2 | patron3:  # varios patrones
        ...
    case _:  # default
        ...
```

````python
comando = "salir"

# match-case requiere Python 3.10+; este código es compatible con Python 3.6+
if comando == "ayuda":
    print("Comandos: iniciar, pausar, salir")
elif comando in ("iniciar", "reanudar"):
    print("Iniciando...")
elif comando == "salir":
    print("Hasta luego")
else:
    print(f"Comando desconocido: {comando}")
````

## Bucles: while y for

**while** repite mientras una condición sea True:
```python
while condicion:
    ...
```

**for** itera sobre una colección:
```python
for elemento in coleccion:
    ...
```

`break` sale del bucle. `continue` salta a la siguiente iteración.

**range(inicio, fin, paso):** genera una secuencia de números.

````python
# while con break
i = 0
while True:
    if i >= 5:
        break
    print(i, end=" ")
    i += 1
print()

# for con range y continue (saltar pares)
for n in range(10):
    if n % 2 == 0:
        continue  # salta los pares
    print(n, end=" ")
print()

# enumerate (índice + valor)
frutas = ["manzana", "pera", "naranja"]
for i, f in enumerate(frutas):
    print(f"  [{i}] {f}")
````

## List Comprehensions

Forma concisa de construir listas:
```python
[expresion for x in iterable if condicion]
```

Equivale a un bucle for + append, pero más legible y rápido.

````python
# Cuadrados de números impares del 1 al 20
cuadrados = [x**2 for x in range(1, 21) if x % 2 != 0]
print(cuadrados)

# Dict comprehension
nombres = ["Ana", "Bob", "Carlos"]
longitudes = {n: len(n) for n in nombres}
print(longitudes)
````

## Definir y llamar funciones

```python
def nombre(param1, param2=default):
    """Docstring: describe qué hace la función"""
    return resultado
```

**Parámetros con valor por defecto:** se usan si el llamante no los proporciona.  
**Retorno múltiple:** en Python puedes devolver una tupla `return a, b`.  
**Anotaciones de tipo:** `def f(x: int) -> str:` (no obligatorias, solo documentan).

````python
def estadisticas(numeros: list):
    """Calcula estadísticas básicas de una lista."""
    if not numeros:
        return {}
    return {
        "min": min(numeros),
        "max": max(numeros),
        "media": sum(numeros) / len(numeros),
        "n": len(numeros)
    }

resultado = estadisticas([3, 1, 4, 1, 5, 9, 2, 6])
for clave in resultado:
    print("clave ",clave, "valor ",resultado[clave])
````

## Lambda, recursión y closures

**Lambda:** función anónima de una sola expresión.
```python
doble = lambda x: x * 2
```

**Recursión:** función que se llama a sí misma. Siempre necesita un **caso base** que detenga la recursión.

**Closure:** función que recuerda variables de su ámbito exterior aunque ese ámbito ya no exista.

````python
# Lambda con sorted
personas = [("Ana", 30), ("Bob", 25), ("Clara", 35)]
orden_edad = sorted(personas, key=lambda p: p[1])
print(orden_edad)

# Recursión: factorial
def factorial(n):
    if n <= 1:     # caso base
        return 1
    return n * factorial(n - 1)

print(factorial(6))  # 720

# Closure
def make_contador(inicio=0):
    cuenta = [inicio]  # lista para poder modificar
    def incrementar():
        cuenta[0] += 1
        return cuenta[0]
    return incrementar

c = make_contador()
print(c(), c(), c())  # 1 2 3
````

## Leer y escribir archivos

Python puede trabajar con archivos de texto con `open()`.

**Modos más comunes:**

|------|-------------|

| `'a'` | Añadir al final |
| `'x'` | Crear nuevo (error si existe) |
| `'rb'`/`'wb'` | Binario |

Siempre usar `with` para que el archivo se cierre automáticamente aunque haya errores.

````python
import os

# Escribir un archivo
ruta = os.path.join("res", "prueba.txt")
with open(ruta, "w", encoding="utf-8") as f:
    f.write("Primera línea\n")
    f.write("Segunda línea\n")
    f.writelines(["Línea 3\n", "Línea 4\n"])

# Leer el archivo
with open(ruta, "r", encoding="utf-8") as f:
    for i, linea in enumerate(f):
        print(f"Línea {i+1}: {linea.strip()}")
````

## Clases y objetos

La **Programación Orientada a Objetos** organiza el código en **clases** (plantillas) y **objetos** (instancias de esas plantillas).

```python
class MiClase:
    atributo_clase = "compartido"  # todos los objetos lo tienen
    
    def __init__(self, param):     # constructor
        self.atributo = param      # atributo de instancia
    
    def metodo(self):              # método
        return self.atributo
```

`self` es la referencia al objeto actual. Debe ser el primer parámetro de todo método.

````python
class Rectangulo:
    """Representa un rectángulo."""
    
    def __init__(self, ancho: float, alto: float):
        self.ancho = ancho
        self.alto = alto
    
    def area(self) -> float:
        return self.ancho * self.alto
    
    def perimetro(self) -> float:
        return 2 * (self.ancho + self.alto)
    
    def __str__(self) -> str:  # representación como texto
        return f"Rectángulo({self.ancho}x{self.alto})"
    
    def __repr__(self) -> str:
        return f"Rectangulo(ancho={self.ancho}, alto={self.alto})"

r = Rectangulo(5, 3)
print(r)                  # usa __str__
print(f"Área: {r.area()}")
print(f"Perímetro: {r.perimetro()}")
````

## Herencia

Una clase **hija** puede heredar atributos y métodos de una clase **padre**:

```python
class Hijo(Padre):
    def __init__(self, ...):
        super().__init__(...)  # llama al constructor del padre
```

**Visibilidad de atributos:**
- `nombre` → público (accesible desde cualquier sitio)
- `_nombre` → protegido (convención: no usar fuera de la clase)
- `__nombre` → privado (Python aplica name mangling)

````python
class Animal:
    def __init__(self, nombre: str, sonido: str):
        self.nombre = nombre
        self._sonido = sonido  # protegido
    
    def hablar(self) -> str:
        return f"{self.nombre} dice: {self._sonido}"

class Perro(Animal):
    def __init__(self, nombre: str, raza: str):
        super().__init__(nombre, "Guau")
        self.raza = raza
    
    def hablar(self) -> str:  # sobreescribe el método
        return f"{self.nombre} ({self.raza}) dice: ¡{self._sonido}!"
    
    def buscar(self) -> str:
        return f"{self.nombre} está buscando..."

perro = Perro("Rex", "Pastor Alemán")
print(perro.hablar())
print(perro.buscar())
print(isinstance(perro, Animal))  # True: es un Animal
````

## Decoradores: @property, @staticmethod, @classmethod

- `@property`: convierte un método en un atributo de solo lectura (getter)
- `@prop.setter`: permite asignar valor con validación
- `@staticmethod`: método que no necesita `self` (no accede al objeto)
- `@classmethod`: recibe la clase como primer parámetro (`cls`)

````python
class Temperatura:
    def __init__(self, celsius: float):
        self._celsius = celsius
    
    @property
    def celsius(self) -> float:
        return self._celsius
    
    @celsius.setter
    def celsius(self, valor: float):
        if valor < -273.15:
            raise ValueError("Temperatura por debajo del cero absoluto")
        self._celsius = valor
    
    @property
    def fahrenheit(self) -> float:  # solo lectura
        return self._celsius * 9/5 + 32
    
    @staticmethod
    def punto_ebullicion_agua() -> float:
        return 100.0

t = Temperatura(25)
print(f"{t.celsius}°C = {t.fahrenheit}°F")
t.celsius = 100
print(f"{t.celsius}°C = {t.fahrenheit}°F")
print("Ebullición agua:", Temperatura.punto_ebullicion_agua())
````

## Clases abstractas y polimorfismo

Una **clase abstracta** define una interfaz (métodos que las clases hijas DEBEN implementar).

El **polimorfismo** significa que objetos de distintas clases pueden usarse de la misma forma si implementan los mismos métodos.

````python
from abc import ABC, abstractmethod

class Figura(ABC):
    @abstractmethod
    def area(self) -> float:
        pass
    
    @abstractmethod
    def perimetro(self) -> float:
        pass
    
    def descripcion(self):
        return f"{type(self).__name__}: área={self.area():.2f}, perímetro={self.perimetro():.2f}"

class Circulo(Figura):
    def __init__(self, radio: float):
        self.radio = radio
    def area(self): return 3.14159 * self.radio ** 2
    def perimetro(self): return 2 * 3.14159 * self.radio

class Cuadrado(Figura):
    def __init__(self, lado: float):
        self.lado = lado
    def area(self): return self.lado ** 2
    def perimetro(self): return 4 * self.lado

figuras = [Circulo(5), Cuadrado(4), Circulo(3)]
for f in figuras:
    print(f.descripcion())  # polimorfismo
````

## Estructuras de datos clásicas

Más allá de listas y diccionarios, existen estructuras especializadas:

| Estructura | Descripción | Módulo Python |
|-----------|-------------|---------------|
| **Stack** (pila) | LIFO: el último en entrar es el primero en salir | `collections.deque` |
| **Queue** (cola) | FIFO: el primero en entrar es el primero en salir | `collections.deque` |
| **Heap** (montículo) | Cola de prioridad: el menor (o mayor) siempre al frente | `heapq` |
| **Array 2D** | Matriz de elementos | lista de listas |
| **Hash Table** | Búsqueda O(1) por clave | `dict` / `set` |
| **Grafo** | Nodos conectados por aristas | `dict` de listas |

````python
from collections import deque
import heapq

# STACK (pila LIFO)
pila = deque()
pila.append("a")   # push
pila.append("b")
pila.append("c")
print("Stack:", list(pila))
print("Pop:", pila.pop())    # quita el último: "c"

# QUEUE (cola FIFO)
cola = deque()
cola.append("primero")
cola.append("segundo")
cola.append("tercero")
print("Queue:", list(cola))
print("Dequeue:", cola.popleft())  # quita el primero

# HEAP (cola de prioridad)
tareas = []
heapq.heappush(tareas, (3, "Tarea C"))
heapq.heappush(tareas, (1, "Tarea A"))
heapq.heappush(tareas, (2, "Tarea B"))
print("Siguiente tarea:", heapq.heappop(tareas))  # (1, "Tarea A")
````

````python
# GRAFO (lista de adyacencia)
grafo = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}

# BFS (Breadth-First Search)
from collections import deque

def bfs(grafo, inicio):
    visitados = []
    cola = deque([inicio])
    visto = {inicio}
    while cola:
        nodo = cola.popleft()
        visitados.append(nodo)
        for vecino in grafo[nodo]:
            if vecino not in visto:
                visto.add(vecino)
                cola.append(vecino)
    return visitados

print("BFS desde A:", bfs(grafo, "A"))
````

5. **Preferir polimorfismo** sobre cadenas de if/elif

## Testing con pytest

Las pruebas automáticas verifican que tu código funciona correctamente.

**Estructura de un test:**

## ¿Qué es NumPy?

**NumPy** (Numerical Python) es la librería fundamental para computación numérica en Python.

¿Por qué NumPy y no listas de Python?
- Las listas de Python son lentas para operaciones matemáticas (cada elemento es un objeto Python)
- NumPy usa arrays de tipo fijo almacenados de forma contigua en memoria → **10-100x más rápido**
- Operaciones vectorizadas: aplicas una operación a TODOS los elementos sin bucle

## ndarray: el array de NumPy

El objeto central de NumPy es el `ndarray` (n-dimensional array).

**Atributos importantes:**
- `.shape` → dimensiones (filas, columnas, ...)
- `.ndim` → número de dimensiones
- `.size` → número total de elementos
- `.dtype` → tipo de dato

````python
import numpy as np

# Crear arrays
a1 = np.array([1, 2, 3, 4, 5])              # 1D
a2 = np.array([[1, 2, 3], [4, 5, 6]])        # 2D (matriz)
a3 = np.zeros((3, 4))                         # 3x4 de ceros
a4 = np.ones((2, 3))                          # 2x3 de unos
a5 = np.arange(0, 20, 2)                      # 0,2,4,...18
a6 = np.linspace(0, 1, 5)                     # 5 puntos entre 0 y 1
a7 = np.eye(3)                                # identidad 3x3

print("a1:", a1)
print("a2:\n", a2)
print("shape a2:", a2.shape, "| ndim:", a2.ndim, "| size:", a2.size)
print("arange:", a5)
print("linspace:", a6)
````

## Indexado y slicing

Muy similar a las listas, pero en múltiples dimensiones:

```python
arr[fila, columna]         # elemento individual
arr[0:3, 1:4]              # submatriz
arr[arr > 5]               # máscara booleana
arr[[0, 2], [1, 3]]        # fancy indexing
```

**IMPORTANTE:** los slices de NumPy son **vistas** (referencias), no copias.
Si modificas el slice, modificas el original. Usa `.copy()` para evitarlo.

````python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Elemento [1,2]:", arr[1, 2])      # 60
print("Fila 0:", arr[0])                 # [10 20 30]
print("Columna 1:", arr[:, 1])           # [20 50 80]
print("Submatriz:\n", arr[:2, 1:])

# Máscara booleana
print("Mayores que 50:", arr[arr > 50])

# Copia segura
subarray = arr[0:2, 0:2].copy()
subarray[0, 0] = 999
print("arr[0,0] no cambió:", arr[0, 0])  # sigue siendo 10
````

## Operaciones matemáticas y broadcasting

**Broadcasting:** NumPy puede operar entre arrays de distintas formas si son compatibles.
Regla: las dimensiones deben ser iguales o una de ellas debe ser 1.

```
(3,4) + (4,)  → válido (la fila se expande)
(3,4) + (3,1) → válido (la columna se expande)
(3,4) + (2,4) → ERROR
```

````python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

# Operación escalar (aplica a todos)
print("x2:\n", a * 2)
print("cuadrado:\n", a ** 2)

# Broadcasting con vector fila
v = np.array([10, 20, 30])
print("Broadcasting + [10,20,30]:\n", a + v)

# Funciones matemáticas (ufuncs)
print("Raíz cuadrada:", np.sqrt(a))
print("Exponencial:", np.exp(a))
````

## Álgebra lineal y np.where

`np.linalg` tiene: `det()`, `inv()`, `eig()`, `solve()`

El operador `@` hace multiplicación de matrices.

`np.where(condicion, valor_si_true, valor_si_false)` es el equivalente vectorizado del ternario.

````python
# Álgebra lineal
A = np.array([[2, 1], [1, 3]])
b = np.array([5, 10])

# Resolver Ax = b
x = np.linalg.solve(A, b)
print("Solución x:", x)
print("Verificación Ax:", A @ x)  # debe ser igual a b

# np.where
arr = np.array([-3, -1, 0, 2, 4, -5, 7])
resultado = np.where(arr >= 0, arr, 0)  # negativos → 0
print("Sin negativos:", resultado)
````

## Correlaciones

La **correlación** mide la relación lineal entre dos variables:

- **0**: sin relación

## ¿Qué es un generador?

Un **generador** produce valores uno a uno bajo demanda, sin calcularlos todos de antemano.

**¿Por qué usarlos?**
- Con 5 millones de registros, cargarlos todos en memoria colapsa el sistema
- Un generador solo mantiene en memoria el elemento actual
- Usa `yield` en lugar de `return`

```python
def mi_generador(n):
    for i in range(n):
        yield i  # pausa aquí, devuelve i, y continúa en la siguiente llamada
```

`next(gen)` avanza el generador al siguiente valor.

````python
# Comparar: lista vs generador
def cuadrados_lista(n):
    return [i**2 for i in range(n)]

def cuadrados_gen(n):
    for i in range(n):
        yield i**2

# La lista crea todos los valores en memoria
lst = cuadrados_lista(5)
print("Lista:", lst)

# El generador produce uno a la vez
gen = cuadrados_gen(5)
print("Generador:", gen)          # no muestra los valores
print("Primero:", next(gen))      # 0
print("Segundo:", next(gen))      # 1
print("Resto:", list(gen))        # [4, 9, 16] (ya consumió 0 y 1)
````

## Generator Expressions

Como las list comprehensions pero con `()` en lugar de `[]`.
No evalúan todos los elementos: son **lazy** (perezosas).

```python
lista = [x**2 for x in range(100)]   # crea 100 elementos en memoria
gen   = (x**2 for x in range(100))   # no crea nada hasta que se pida
```

````python
# Generator expression
cuadrados = (x**2 for x in range(10))
print(sum(cuadrados))  # consume el generador calculando la suma

# Pipeline de generadores: leer CSV línea a línea sin cargar todo
import os
ruta = os.path.join("res", "titanic.csv")

with open(ruta) as f:
    # Pipeline encadenado
    lineas = (l.rstrip() for l in f)                       # leer líneas
    filas = (l.split(",") for l in lineas)                 # dividir
    cabecera = next(filas)                                 # extraer cabecera
    idx_survived = cabecera.index("Survived")
    idx_pclass   = cabecera.index("Pclass")
    
    # Solo clase 1
    clase1 = (f for f in filas if len(f) > idx_pclass and f[idx_pclass] == "1")
    sobrevivientes_clase1 = sum(int(f[idx_survived]) for f in clase1 if f[idx_survived].isdigit())

print(f"Sobrevivientes en clase 1: {sobrevivientes_clase1}")
````

## Generadores para entrenamiento de modelos

En Deep Learning, los datasets pueden tener millones de imágenes o líneas.
Se usan generadores (o `tf.keras.utils.Sequence`) para cargar los datos en **lotes (batches)**.

Keras tiene `ImageDataGenerator` para imágenes:
- Carga imágenes en batches desde un directorio
- Aplica **data augmentation** en tiempo real (rotaciones, zooms, flips...)

````python
# Ejemplo conceptual de generador de batches
import numpy as np

def batch_generator(data, labels, batch_size=32):
    """Generador que produce batches de datos."""
    n = len(data)
    indices = np.arange(n)
    while True:  # bucle infinito: Keras lo controla con steps_per_epoch
        np.random.shuffle(indices)
        for start in range(0, n, batch_size):
            batch_idx = indices[start:start + batch_size]
            yield data[batch_idx], labels[batch_idx]

# Simular uso
X = np.random.rand(1000, 10)
y = np.random.randint(0, 2, 1000)

gen = batch_generator(X, y, batch_size=64)
X_batch, y_batch = next(gen)
print(f"Batch shape: X={X_batch.shape}, y={y_batch.shape}")
````

## NLTK: Natural Language Toolkit

**NLTK** es la librería estándar para NLP en Python.
Permite analizar texto: tokenizar, contar palabras, encontrar sinónimos, analizar sentimientos...

**Conceptos clave:**
- **Corpus**: colección de textos (p.ej. todas las obras de Shakespeare)
- **Token**: unidad mínima de texto (generalmente una palabra)
- **Tokenización**: dividir un texto en tokens
- **Stopwords**: palabras muy comunes sin carga semántica ("the", "a", "of"...)
- **Stemming**: reducir palabras a su raíz ("running" → "run")
- **Lematización**: obtener la forma canónica ("ran" → "run")

````python
import nltk

# Descargar datos necesarios (solo la primera vez)
nltk.download("gutenberg", quiet=True)
nltk.download("punkt", quiet=True)
nltk.download("stopwords", quiet=True)

# Cargar texto de Alicia en el País de las Maravillas
alice = nltk.corpus.gutenberg.words("carroll-alice.txt")
print(f"Total palabras: {len(alice)}")
print("Primeras 10:", list(alice[:10]))
````

````python
import string

# Frecuencia de palabras (filtrando puntuación y stopwords)
puntuacion = set(string.punctuation)
stopwords = set(nltk.corpus.stopwords.words("english"))

alice_filtrado = [
    w.lower() for w in alice
    if w.lower() not in stopwords
    and w.lower() not in puntuacion
    and w.isalpha()
]

fd = nltk.FreqDist(alice_filtrado)
print("Palabras más frecuentes:")
for palabra, count in fd.most_common(10):
    print(f"  {palabra}: {count}")
````

## Bigramas, sinónimos y análisis de sentimiento

- **Bigramas**: pares de palabras que aparecen juntas con frecuencia
- **WordNet**: base de datos léxica con sinónimos, antónimos y relaciones entre palabras
- **Análisis de sentimiento**: clasificar texto como positivo, negativo o neutral

````python
# Bigramas
bigrams = nltk.FreqDist(nltk.bigrams(alice_filtrado))
print("Bigramas más comunes:")
for bigrama, count in bigrams.most_common(5):
    print(f"  {bigrama[0]} + {bigrama[1]}: {count}")
````

````python
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
from nltk.corpus import wordnet

# Sinónimos de "happy"
sinonimos = set()
for syn in wordnet.synsets("happy"):
    for lemma in syn.lemmas():
        sinonimos.add(lemma.name())
print("Sinónimos de happy:", sinonimos)

# Stemming
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
palabras = ["running", "ran", "runs", "easily", "fairly"]
for p in palabras:
    print(f"  {p} → {stemmer.stem(p)}")
````

## scikit-learn: machine learning en Python

**scikit-learn** es la librería estándar de Machine Learning en Python.
Incluye algoritmos de clasificación, regresión, clustering, y reducción de dimensionalidad.

**Flujo típico de ML:**
1. **Cargar datos**
2. **Explorar y preprocesar** (limpiar, normalizar...)
3. **Dividir** en entrenamiento y test
4. **Entrenar** el modelo
5. **Evaluar** el rendimiento
6. **Predecir** con nuevos datos

## Dataset Digits

Este notebook usa el dataset `digits` de sklearn:
- 1797 imágenes de dígitos escritos a mano
- Cada imagen es 8x8 píxeles en escala de grises
- La tarea: clasificar qué dígito es (0-9)

````python
from sklearn import datasets
import matplotlib.pyplot as plt
import numpy as np

# Cargar dataset
digits = datasets.load_digits()

print("Shape imágenes:", digits.images.shape)
print("Shape datos 1D:", digits.data.shape)
print("Clases:", digits.target_names)

# Visualizar algunas imágenes
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for ax, img, label in zip(axes.flat, digits.images, digits.target):
    ax.imshow(img, cmap="gray_r")
    ax.set_title(f"Dígito: {label}")
    ax.axis("off")
fig.suptitle("Ejemplos del dataset Digits")
plt.tight_layout()
plt.show()
````

````python
from sklearn.model_selection import train_test_split
from sklearn import svm, metrics

# Aplanar imágenes: de (8,8) a (64,)
n_muestras = len(digits.images)
X = digits.images.reshape(n_muestras, -1)
y = digits.target

# Dividir: 95% entrenamiento, 5% test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.05, random_state=42
)

print(f"Entrenamiento: {len(X_train)} muestras")
print(f"Test: {len(X_test)} muestras")

# Entrenar SVM
modelo = svm.SVC(gamma=0.001, C=100)
modelo.fit(X_train, y_train)

# Predecir y evaluar
predicciones = modelo.predict(X_test)
precision = metrics.accuracy_score(y_test, predicciones)
print(f"\nPrecisión: {precision:.1%}")
````

````python
# Visualizar predicciones
fig, axes = plt.subplots(1, 8, figsize=(15, 3))
for ax, img, pred, real in zip(axes, X_test[:8], predicciones[:8], y_test[:8]):
    ax.imshow(img.reshape(8, 8), cmap="gray_r")
    color = "green" if pred == real else "red"
    ax.set_title(f"Pred:{pred}", color=color)
    ax.axis("off")
plt.suptitle("Verde=correcto, Rojo=error")
plt.tight_layout()
plt.show()

print("Informe de clasificación:")
print(metrics.classification_report(y_test, predicciones))
````

## OpenCV: visión por computador

**OpenCV** (Open Computer Vision) es la librería más usada para procesamiento de imágenes y visión por computador.

**Conceptos clave:**
- Una imagen digital es una **matriz de píxeles**
- Cada píxel tiene valores de color: en escala de grises 0-255; en color RGB/BGR tres valores
- **OpenCV usa BGR** (Blue-Green-Red), no RGB
- Operaciones comunes: recortar, redimensionar, detectar bordes, detectar objetos

**Filtros de bordes:**
- **Sobel**: detecta bordes en dirección horizontal o vertical
- **Laplacian**: detecta bordes en todas las direcciones
- **Canny**: detector de bordes robusto (combina Sobel + supresión + histéresis)

````python
try:
    import cv2
    print("OpenCV versión:", cv2.__version__)
except ImportError:
    print("OpenCV no instalado. Instalar con: pip install opencv-python")

import numpy as np
import matplotlib.pyplot as plt
````

````python
# Crear una imagen sintética para demostrar operaciones
import numpy as np
import matplotlib.pyplot as plt

# Imagen de ejemplo: gradiente con formas geométricas
img = np.zeros((200, 200), dtype=np.uint8)
img[50:150, 50:150] = 150  # cuadrado gris
img[80:120, 80:120] = 255  # cuadrado blanco interior

# Simular detección de bordes manualmente (gradiente de Sobel simplificado)
gx = np.diff(img.astype(float), axis=1, prepend=0)
gy = np.diff(img.astype(float), axis=0, prepend=0)
borde = np.sqrt(gx**2 + gy**2)
borde = (borde / borde.max() * 255).astype(np.uint8)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(img, cmap="gray"); axes[0].set_title("Original"); axes[0].axis("off")
axes[1].imshow(borde, cmap="gray"); axes[1].set_title("Bordes"); axes[1].axis("off")

# Histograma de intensidades
axes[2].hist(img.ravel(), bins=50, color="steelblue", edgecolor="navy")
axes[2].set_title("Histograma de píxeles")
axes[2].set_xlabel("Intensidad (0-255)")

fig.tight_layout()
plt.show()
````
