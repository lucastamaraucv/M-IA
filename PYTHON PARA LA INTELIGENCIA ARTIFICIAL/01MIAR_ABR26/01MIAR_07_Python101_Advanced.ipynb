## Array

<div style="background-color:#e7f5ff;border-left:6px solid #1c7ed6;padding:12px 16px;border-radius:6px;margin:6px 0;"><b style="color:#1864ab;font-size:14px;">TEORIA &middot; Array: la estructura de datos mas basica</b>

Un **array** es una **secuencia de elementos del mismo tipo guardados en memoria contigua**. Es la estructura mas eficiente para acceso por indice.

#### Caracteristicas

| Operacion | Coste | Por que |
|---|---|---|
| Acceso por indice `a[i]` | **O(1)** | Calcula la direccion: `base + i * tam` |
| Buscar valor | O(n) | Hay que recorrer |
| Insertar al final | O(1) amortizado (en `list`) | Si hay capacidad |
| Insertar en medio | O(n) | Hay que desplazar elementos |
| Borrar | O(n) | Igual: desplazamiento |

#### En Python

Python NO tiene arrays clasicos por defecto: tiene **`list`**, que es un array dinamico (se redimensiona solo). Para arrays homogeneos eficientes existe el modulo `array` o, mejor, **`numpy.ndarray`**.

```python
# list (dinamica, tipos mixtos)
nums = [1, 2, 3]
nums.append(4)        # O(1) amortizado
nums[0]               # O(1) acceso

# array (homogeneo, mas compacto)
import array
a = array.array('i', [1, 2, 3])   # 'i' = int de 4 bytes

# numpy (la herramienta de IA)
import numpy as np
v = np.array([1, 2, 3])
```

#### Cuando usarlo?

- Cuando necesitas acceso aleatorio rapido por indice.
- Cuando el tamano es relativamente estable o crece al final.
- **NO** cuando insertas/borras mucho en el medio (mejor lista enlazada o deque).

## Lista enlazada (Linked list)

<div style="background-color:#e7f5ff;border-left:6px solid #1c7ed6;padding:12px 16px;border-radius:6px;margin:6px 0;"><b style="color:#1864ab;font-size:14px;">TEORIA &middot; Lista enlazada: nodos conectados por punteros</b>

Una **lista enlazada** es una secuencia de **nodos** donde cada nodo guarda:

- Un **valor**.
- Una **referencia** (puntero) al siguiente nodo.

```
[1|*]-->[2|*]-->[3|*]-->[4|None]
```

#### Comparacion con array

| Operacion | Array | Lista enlazada |
|---|---|---|
| Acceso por indice | **O(1)** | O(n) (hay que recorrer) |
| Insertar al inicio | O(n) | **O(1)** |
| Insertar al final | O(1) amortizado | O(n) (o O(1) si tienes puntero al final) |
| Insertar en medio (con referencia al nodo) | O(n) | **O(1)** |
| Memoria | Compacta | Dispersa (mas overhead por puntero) |
| Cache CPU | Buen aprovechamiento | Malo (saltos en memoria) |

#### Variantes

- **Simple**: cada nodo apunta solo al siguiente.
- **Doble**: cada nodo apunta al siguiente Y al anterior. Permite recorrer en ambas direcciones.
- **Circular**: el ultimo nodo apunta al primero.

#### En Python

Python NO tiene lista enlazada en la libreria estandar (la `list` es array dinamico). Si necesitas comportamiento similar, usa **`collections.deque`** que esta implementada como lista doblemente enlazada y ofrece `append`, `appendleft`, `pop`, `popleft` en O(1).

## Pila (Stack)

<div style="background-color:#e7f5ff;border-left:6px solid #1c7ed6;padding:12px 16px;border-radius:6px;margin:6px 0;"><b style="color:#1864ab;font-size:14px;">TEORIA &middot; Pila (Stack): el ultimo en entrar es el primero en salir (LIFO)</b>

Una **pila** sigue el principio **LIFO** (*Last In, First Out*): el ultimo elemento agregado es el primero en salir. Como una pila de platos: solo puedes coger el de arriba.

#### Operaciones

| Operacion | Que hace | Coste |
|---|---|---|
| `push(x)` | Apila `x` arriba | O(1) |
| `pop()` | Quita y devuelve el de arriba | O(1) |
| `peek()` / `top()` | Mira el de arriba sin quitarlo | O(1) |
| `is_empty()` | True si esta vacia | O(1) |

#### En Python

Tres opciones:

```python
# 1. list (la mas comun, simple)
pila = []
pila.append(1)        # push
pila.append(2)
x = pila.pop()        # pop -> 2
arriba = pila[-1]     # peek -> 1

# 2. collections.deque (mejor si haces muchos pop/push)
from collections import deque
pila = deque()
pila.append(1)
pila.pop()

# 3. queue.LifoQueue (thread-safe, util en concurrencia)
from queue import LifoQueue
```

#### Casos de uso reales

- **Funciones**: la pila de llamadas (call stack) que ves en los tracebacks.
- **Deshacer/Rehacer** en editores (undo/redo).
- **Validar parentesis balanceados**: `({[]})` -> apila al abrir, desapila al cerrar.
- **DFS** (busqueda en profundidad) en grafos.
- **Evaluar expresiones** con notacion polaca inversa.

* La operación *pop* siempre tiene coste bajo.

## Cola (Queue)

<div style="background-color:#e7f5ff;border-left:6px solid #1c7ed6;padding:12px 16px;border-radius:6px;margin:6px 0;"><b style="color:#1864ab;font-size:14px;">TEORIA &middot; Cola (Queue): el primero en entrar es el primero en salir (FIFO)</b>

Una **cola** sigue el principio **FIFO** (*First In, First Out*): el primer elemento agregado es el primero en salir. Como la cola del supermercado.

#### Operaciones

| Operacion | Que hace | Coste |
|---|---|---|
| `enqueue(x)` | Agrega al final | O(1) |
| `dequeue()` | Saca y devuelve el primero | O(1) |
| `peek()` | Mira el primero sin sacarlo | O(1) |
| `is_empty()` | True si esta vacia | O(1) |

#### En Python: NO uses `list` para colas

```python
# MAL: muy lento porque pop(0) es O(n)
cola = []
cola.append(1)
cola.pop(0)        # O(n) -> desplaza todos los elementos!

# BIEN: deque tiene popleft en O(1)
from collections import deque
cola = deque()
cola.append(1)        # enqueue
cola.append(2)
x = cola.popleft()    # dequeue -> 1
```

#### Casos de uso reales

- **Colas de tareas / mensajes**: procesar trabajos en orden de llegada.
- **BFS** (busqueda en anchura) en grafos.
- **Buffer de eventos**: teclas presionadas, paquetes de red.
- **Colas de impresion**.
- **Productores-consumidores** en concurrencia (`queue.Queue` es thread-safe).

#### Variantes

- **Cola doble (deque)**: permite agregar y sacar por ambos extremos. `collections.deque`.
- **Cola circular**: tamano fijo; al llenarse, se sobrescribe la cabeza.
- **Cola de prioridad**: el orden NO es por llegada sino por una prioridad. Lo veras en la siguiente seccion.

## Cola de prioridad (heaps)

<div style="background-color:#e7f5ff;border-left:6px solid #1c7ed6;padding:12px 16px;border-radius:6px;margin:6px 0;"><b style="color:#1864ab;font-size:14px;">TEORIA &middot; Cola de prioridad (Heap): el mas urgente primero</b>

Una **cola de prioridad** devuelve los elementos en orden de **prioridad**, no en orden de llegada. Internamente se implementa con un **heap** (monticulo): un arbol binario donde cada padre es menor (min-heap) o mayor (max-heap) que sus hijos.

#### Operaciones

| Operacion | Que hace | Coste |
|---|---|---|
| `push(x)` | Inserta `x` | O(log n) |
| `pop()` | Saca el de mayor (o menor) prioridad | O(log n) |
| `peek()` | Mira el primero sin sacarlo | O(1) |

#### En Python: modulo `heapq`

`heapq` implementa un **min-heap** sobre una lista:

```python
import heapq

heap = []
heapq.heappush(heap, 3)
heapq.heappush(heap, 1)
heapq.heappush(heap, 2)

print(heapq.heappop(heap))   # 1 (el menor)
print(heapq.heappop(heap))   # 2
print(heapq.heappop(heap))   # 3
```

#### Como hacer un MAX-heap?

`heapq` solo es min-heap. Truco: **negar** los valores.

```python
heapq.heappush(heap, -valor)
mayor = -heapq.heappop(heap)
```

#### Heap con elementos complejos (prioridad + dato)

Apila **tuplas** `(prioridad, dato)`. Python compara por el primer elemento.

```python
tareas = []
heapq.heappush(tareas, (2, 'limpiar'))
heapq.heappush(tareas, (1, 'urgente: bug'))
heapq.heappush(tareas, (3, 'reunion'))

while tareas:
    prio, t = heapq.heappop(tareas)
    print(prio, t)
# 1 urgente: bug
# 2 limpiar
# 3 reunion
```

#### Casos de uso reales

- **Algoritmo de Dijkstra** (caminos mas cortos).
- **Algoritmo A*** (busqueda informada).
- **Schedulers**: ejecutar tareas por prioridad.
- **Top-K**: encontrar los `k` mayores/menores en un stream.

## Tabla Hash

<div style="background-color:#e7f5ff;border-left:6px solid #1c7ed6;padding:12px 16px;border-radius:6px;margin:6px 0;"><b style="color:#1864ab;font-size:14px;">TEORIA &middot; Tabla hash: busqueda en tiempo casi constante</b>

Una **tabla hash** asocia **claves** con **valores** y permite consultar el valor de una clave en **O(1) promedio**. En Python es el `dict`.

#### Como funciona por dentro?

1. Se calcula `hash(clave)`: una funcion que convierte la clave en un entero.
2. Se reduce ese entero al rango de un array interno: `indice = hash % tamano`.
3. Se guarda el par `(clave, valor)` en ese indice.

```
hash('Ana') = 9028273...
9028273 % 8 = 5
-> array[5] = ('Ana', 30)
```

#### Colisiones

Dos claves distintas pueden caer en el mismo indice. Las dos estrategias clasicas:

- **Encadenamiento (chaining)**: cada celda guarda una lista de pares.
- **Direccionamiento abierto**: si la celda esta ocupada, prueba la siguiente.

Python usa una variante de direccionamiento abierto.

#### Costes

| Operacion | Promedio | Peor caso |
|---|---|---|
| Insertar | **O(1)** | O(n) (muchas colisiones) |
| Buscar | **O(1)** | O(n) |
| Borrar | **O(1)** | O(n) |

#### Que tipos pueden ser claves?

Solo objetos **hashables** (inmutables): `int`, `float`, `str`, `tuple` (de hashables), `frozenset`, etc.

**NO** se pueden usar como clave: `list`, `dict`, `set` (mutables).

```python
d = {}
d[(1, 2)] = 'ok'      # tupla -> ok
# d[[1, 2]] = 'no'    # TypeError: unhashable
```

#### Operaciones tipicas en `dict`

```python
d = {'a': 1, 'b': 2}
d['c'] = 3            # insertar/actualizar
d.get('x', 0)         # default si no existe
'a' in d              # True (busqueda O(1))
del d['a']            # borrar
for k, v in d.items():
    print(k, v)
```

#### Set: la primita del dict

`set` es una tabla hash sin valores (solo claves). Util para:
- Quitar duplicados: `set([1, 2, 2, 3]) -> {1, 2, 3}`.
- Test de pertenencia rapido: `x in mi_set` es O(1).

## Grafo

<div style="background-color:#e7f5ff;border-left:6px solid #1c7ed6;padding:12px 16px;border-radius:6px;margin:6px 0;"><b style="color:#1864ab;font-size:14px;">TEORIA &middot; Grafo: nodos conectados por aristas</b>

Un **grafo** es un conjunto de **nodos (vertices)** conectados por **aristas (edges)**. Modelan relaciones: redes sociales, mapas de carreteras, dependencias entre tareas, etc.

#### Tipos

| Tipo | Caracteristicas | Ejemplo |
|---|---|---|
| **No dirigido** | Aristas sin direccion | Amistades en Facebook |
| **Dirigido** (digrafo) | Aristas con direccion | Seguidores en Twitter, web links |
| **Ponderado** | Las aristas tienen peso | Distancia entre ciudades |
| **Ciclico / Aciclico** | Tiene ciclos o no (DAG) | DAG: dependencias de tareas |

#### Representaciones tipicas

**1. Lista de adyacencia** (la mas comun, eficiente en memoria):

```python
grafo = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A'],
    'D': ['B'],
}
```

**2. Matriz de adyacencia** (`n x n`):

```
    A  B  C  D
A [ 0  1  1  0 ]
B [ 1  0  0  1 ]
C [ 1  0  0  0 ]
D [ 0  1  0  0 ]
```

| | Lista de adyacencia | Matriz de adyacencia |
|---|---|---|
| Memoria | O(V + E) | O(V^2) |
| Saber si A conecta con B | O(grado(A)) | **O(1)** |
| Iterar vecinos | **Eficiente** | O(V) |

#### Algoritmos clasicos

- **BFS** (busqueda en anchura): usa **cola**. Encuentra el camino mas corto en aristas.
- **DFS** (busqueda en profundidad): usa **pila** o recursion. Util para detectar ciclos.
- **Dijkstra**: caminos mas cortos en grafos ponderados (heap).
- **Topological sort**: ordenar tareas con dependencias (DAG).

# Software Craftsmanship

<div style="background-color:#e7f5ff;border-left:6px solid #1c7ed6;padding:12px 16px;border-radius:6px;margin:6px 0;"><b style="color:#1864ab;font-size:14px;">TEORIA &middot; Software Craftsmanship: por que importa el codigo limpio</b>

**Software Craftsmanship** es la idea de tratar el desarrollo como un **oficio**: con orgullo, calidad y mejora continua. El movimiento se popularizo con libros como *Clean Code* (Robert C. Martin) y *The Pragmatic Programmer*.

#### Por que importa?

- **Pasamos mas tiempo leyendo codigo que escribiendolo** (proporcion 10:1 segun Uncle Bob). Codigo legible = mas productividad.
- **Los bugs viven en el codigo confuso.** Codigo claro = menos errores.
- **El equipo cambia.** Tu codigo debe ser mantenible por alguien que llega manana.
- **Refactorizar codigo malo es caro.** El "deuda tecnica" se paga con intereses.

#### Reglas basicas (resumen del libro Clean Code)

1. **Nombres significativos**: `n` -> `numero_alumnos`. Variable, funcion y clase deben decir QUE son.
2. **Funciones cortas**: idealmente menos de 20 lineas. Una sola responsabilidad.
3. **Argumentos pocos**: 0-2 ideales, 3 ya es mucho. Mas de 3 -> agrupa en objeto.
4. **Sin duplicacion (DRY)**: *Don't Repeat Yourself*. Cada conocimiento, una sola vez.
5. **Comentarios solo para el porque**: el codigo dice QUE hace. El comentario dice POR QUE.
6. **Manejo de errores explicito**: no silencies excepciones, no ignores codigos de error.
7. **Tests**: codigo sin tests es legacy desde el dia 1.

#### Principios SOLID (orientacion a objetos)

| Letra | Principio | Idea |
|---|---|---|
| **S** | Single Responsibility | Una clase, una razon para cambiar |
| **O** | Open/Closed | Abierto a extension, cerrado a modificacion |
| **L** | Liskov Substitution | Subclases deben ser intercambiables con su padre |
| **I** | Interface Segregation | Interfaces pequenas, especificas |
| **D** | Dependency Inversion | Depende de abstracciones, no de implementaciones |

#### El boy scout rule

> "Deja el codigo mas limpio de como lo encontraste."

Cada vez que tocas un fichero, mejoralo un poco: renombra una variable, extrae una funcion, anade un test.

- Funcion con mas de 50 lineas. Probablemente hace varias cosas.

- Funcion con mas de 4 argumentos. Difícil de usar y probar.

| Sin tests | Con tests |
|---|---|
| Probar a mano cada cambio | Ejecutas `pytest` y todo se valida |

#### La piramide de testing

```
            /\
           /  \   E2E (pocos, lentos, frágiles)
          /----\
         /      \  Integracion
        /--------\
       /          \ Unitarios (muchos, rapidos)
      /____________\
```

- **Unitarios**: prueban una funcion / clase aislada. Rapidos (ms). Muchos.
- **Integracion**: prueban como interactuan varios componentes (DB, API).
- **End-to-end (E2E)**: prueban el sistema completo desde la UI o la API publica. Lentos.

**Regla del 70/20/10**: 70% unitarios, 20% integracion, 10% E2E.

#### TDD: Test-Driven Development

Ciclo **rojo-verde-refactor**:

1. **Rojo**: escribe un test que falla.
2. **Verde**: escribe el codigo MINIMO para que pase.
3. **Refactor**: limpia sin romper los tests.

Beneficios: te obliga a pensar la API antes de implementar, te da cobertura desde el inicio, te lleva a un diseno mas modular.

## ¿Qué es el software testing? 

* Testing es la actividad de verificación de que el software funciona correctamente.
* Puede ser manual o automático.

#### Testing manual

* Una persona prueba manualmente que una aplicación hace lo que se espera.
    * Ejecución de *casos de prueba* especificados en un documento.
    * Realización de testing *exploratorio*.

Ejemplo de caso de prueba para una calculadora:

* Introducir los valores 4 y 5, ejecutar la acción multiplicar y comprobar que el resultado es 20.

#### Testing automático

* Los *casos de prueba* pueden automatizarse a través de código fuente.
    * Implementación de un "programa que prueba otro programa".


* Los tests automatizados son mucho más eficientes que los tests manuales ya que se pueden ejecutar las veces que queramos sin la intervención de un ser humano.

De aquí en adelante, se hace referencia, salvo que se especifique lo contrario, únicamente a tests automatizados, ya que son los que como programadores más valor nos van a aportar.

## Tipos de tests

Uno de los principales objetivos del testing es la **prevención** de bugs.

* *Tests unitarios*: verifican módulos del software de manera aislada.
* *Tests de aceptación*: verifican funcionalidad esperada por el usuario.
* *Tests de interfaz gráfica (UI)*: verifican que la aplicación, a través de su interfaz gráfica, funciona correctamente.

Además, también podemos probar el software de manera manual y exploratoria para **localización** de bugs que se nos hayan podido escapar.

<img src="img/testing/testing_pyramid.png" width="700">

Otra terminología que se puede encontrar dentro del campo del software testing:

* *Tests de integración*: prueban varios módulos unitarios de manera integrada.
* *Tests funcionales*: prueban funcionalidad que se espera de la aplicación. Es un término similar a tests de aceptación.
* *Tests no funcionales*: prueban requisitos no funcionales, como rendimiento, escalabilidad, etc.

## Estructura de un test automatizado

Ejemplo: comprobación de que una función 'eliminar_duplicados' elimina correctamente los duplicados de una lista.

#### Setup (test fixture)

* En este paso se crean de todos los objetos necesarios para ejecutar el test.

Ejemplo: creación de la lista [1, 2, 2, 2, 3, 4, 4]

#### Acción

* Invocación del código que queremos probar.

Ejemplo: invocación de la función 'eliminar_duplicados'

#### Verificación resultado

* Comprobación de que la acción que hemos invocado ha producido el resultado esperado.

Ejemplo: comprobación que la lista resultante tiene los elementos esperados, en este caso [1, 2, 3, 4].

#### Tear down

* Cada test automatizado debe dejar el sistema en el mismo estado que estaba antes de ejecutar el test.
* Así se evita que unos tests afecten los resultados de otros tests.
* En el paso de 'tear down' se eliminan de todos los recursos persistentes creados por el test (por ejemplo, ficheros o registros en una BD).

Ejemplo: no aplica en el ejemplo de la lista.

## Beneficios del testing automático

* Más eficiente que el testing manual, lo que conlleva una reducción de costes.
* Reducción del número de bugs (prevención vs localización).
* Incremento de **calidad**: permiten modificar el código para hacerlo más legible sin que tengamos miedo de introducir bugs.
   * Los tests proporcionan una **red de seguridad**.
* Facilitan la comprensión del sistema.
    * Los tests proporcionan **especificaciones ejecutables** del comportamiento esperado.
    * Estas especificaciones nos sirven como **ejemplos de código** que nos dicen cómo se debe usar el sistema.

## Test-Driven Development (TDD)

* Un proceso de desarrollo en el cual los tests se crean antes que el código del sistema (*test-first*).
* El código que desarrollamos va dirigido por los tests: cada test que añadimos representa una evolución del sistema.

Proceso:

1. Añadir test.
2. Ejecutar test. Debe fallar porque el código para que el test pase aun no existe.
3. Implementar código para que pase el test.
4. Ejecutar tests. Todos los tests implementados deberían pasar.
5. Mejorar la calidad del código (refactoring).
6. Repetir.

Más información [aquí](https://en.wikipedia.org/wiki/Test-driven_development).

## El testing en el ciclo de vida del software

Dos ciclos de vida del software:
   
   * Desarrollo en cascada (waterfall).
   * Desarrollo ágil.

#### Ciclo en cascada

* Herencia de la revolución industrial.
  * Emula el desarrollo de productos físicos (casas, coches, puentes, etc.) en fases que se ejecutan de manera secuencial.
  * El testing es una etapa que se ejecuta únicamente al final del proceso.


* Los proyectos tienen una fecha de inicio y una fecha de fin, donde el alcance está prefijado de antemano y se elabora una planificación detallada.

* Fases:
  1. *Analisis*: análisis de los requisitos. Elaboración de uno o más documentos con las especificaciones de los requisitos que deberá cumplir el producto final.
  2. *Diseño*: diseño o especificación de la solución a un alto nivel de abstracción.
  3. *Implementación*: etapa de construcción del producto demandado por el cliente.
  4. *Test*: etapa de pruebas donde se verifica que el producto construido cumple con los requisitos especificados.
  5. *Mantenimiento*: fase de mantenimiento del producto, donde se realiza cualquier operación necesaria tras la entrega a cliente.

<img src="img/testing/waterfall.png" width="700">

**Problemas del ciclo en cascada**

* Intenta introducir certidumbre en un proceso altamente incierto como es el desarrollo de software.
  * Se puede tratar de planificar un proyecto de software en detalle, pero décadas de proyectos fracasados han demostrado que estas planificaciones detalladas nunca se cumplen.
* No da soporte a introducir cambios a mitad del proceso.
  * En la construcción de productos físicos, esto no es un problema, pero sí lo es en software que, por su naturaleza intangible, es fácil de cambiar.
* Clientes insatisfechos que no reciben el producto que esperaban.
* Centrado exclusivamente en el cumplimiento de plazos y contratos, no en la entrega de valor al cliente.
  * Orientado a proyecto, no a calidad del producto.

#### Desarrollo de software ágil

* Basado en **iteraciones** y en entregas frecuentes a cliente.
* El producto se construye de manera **incremental**.
* El cliente puede ver versiones funcionales del producto frecuentemente, proponer cambios y dar feedback, lo que incrementa el aporte de valor.
* En cada iteración se construye lo más prioritario.
* Centrado en la **excelencia técnica**, ya que no se pueden hacer entregas frecuentes si el código no es de calidad.
* Dado que la **calidad** es tan importante, el **testing** es una actividad que se realiza constantemente, no como una fase al final del desarrollo.
* Se suele realizar cierto nivel de planificación, pero a más corto plazo y es mucho más cambiante que en el modelo en cascada.
* La base es establecer una relación de **confianza** con el cliente.

Metodología: eXtreme Programming (XP).

````python
# Ejercicio 2

import heapq

def heapsort(lista):
    cola_prioridad = []
  
    for elemento in lista:
         heapq.heappush(cola_prioridad, elemento) 
        
    lista_ordenada = []
    while len(cola_prioridad) > 0:
        elemento_extraido_de_cola = heapq.heappop(cola_prioridad)
        lista_ordenada.append(elemento_extraido_de_cola)
        
    return lista_ordenada


print(heapsort([6,10,1,4,9,16,2,5]))
````

````python
# Ejercicio 4

from collections import deque

def match(c1, c2):
    return (c1 == '(' and c2 == ')') or (c1 == '[' and c2 == ']')

def esta_balanceado(s):
    pila = deque()
    for c in s:
        if c == '(' or c == '[':
            pila.append(c)
        elif c == ')' or c == ']':
            if len(pila) == 0 or not match(pila.pop(), c):
                return False
    return len(pila) == 0


print(esta_balanceado('(aa(c))bb[a]'))
print(esta_balanceado('([b])(aa)'))
print(esta_balanceado('dd((abc[a)]'))
print(esta_balanceado('[(a])'))
````
