## Metodos y funciones usados en esta sesion

- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".
- super() — llama a la implementacion de la clase padre (usado en herencia).
- isinstance(obj, Clase) — comprueba el tipo (o herencia) de un objeto.

## Codigo de ejemplo

```python
class Animal:
    def __init__(self, nombre: str, sonido: str):
        self.nombre = nombre
        self._sonido = sonido

    def hablar(self) -> str:
        return f"{self.nombre} dice: {self._sonido}"

class Perro(Animal):
    def __init__(self, nombre: str, raza: str):
        super().__init__(nombre, "Guau")
        self.raza = raza

    def hablar(self) -> str:
        return f"{self.nombre} ({self.raza}) dice: {self._sonido}!"

    def buscar(self) -> str:
        return f"{self.nombre} esta buscando..."

perro = Perro("Rex", "Pastor Aleman")
print(perro.hablar())
print(perro.buscar())
print(isinstance(perro, Animal))
```
