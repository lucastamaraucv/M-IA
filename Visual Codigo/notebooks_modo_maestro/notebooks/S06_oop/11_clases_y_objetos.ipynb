## Metodos y funciones usados en esta sesion

- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".

## Codigo de ejemplo

```python
class Rectangulo:
    def __init__(self, ancho: float, alto: float):
        self.ancho = ancho
        self.alto = alto

    def area(self) -> float:
        return self.ancho * self.alto

    def perimetro(self) -> float:
        return 2 * (self.ancho + self.alto)

    def __str__(self) -> str:
        return f"Rectangulo({self.ancho}x{self.alto})"

r = Rectangulo(5, 3)
print(r)
print(f"Area: {r.area()}")
print(f"Perimetro: {r.perimetro()}")
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Define la clase `Rectangulo`:
    - Define la funcion `__init__(self, ancho: float, alto: float)`:
        - Asigna a `self.ancho`: `ancho`
        - Asigna a `self.alto`: `alto`
    - Define la funcion `area(self)`:
        - **Devuelve** `self.ancho * self.alto`.
    - Define la funcion `perimetro(self)`:
        - **Devuelve** `2 * (self.ancho + self.alto)`.
    - Define la funcion `__str__(self)`:
        - **Devuelve** `f'Rectangulo({self.ancho}x{self.alto})'` — *f-string — interpolacion de expresiones dentro de un string con f'...{expr}...'.*.
- Asigna a `r`: `Rectangulo(5, 3)`
