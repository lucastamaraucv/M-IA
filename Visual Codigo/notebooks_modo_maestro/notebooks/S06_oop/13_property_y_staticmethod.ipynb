## Metodos y funciones usados en esta sesion

- @property — convierte un metodo en un atributo de solo lectura (se accede sin parentesis).
- raise Excepcion(...) — lanza una excepcion manualmente.
- @staticmethod — metodo que no recibe self, vive en la clase solo por organizacion.
- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".

## Codigo de ejemplo

```python
class Temperatura:
    def __init__(self, celsius: float):
        self._celsius = celsius

    @property
    def celsius(self) -> float:
        return self._celsius

    @celsius.setter
    def celsius(self, valor: float):
        if valor < -273.15:
            raise ValueError("Bajo cero absoluto")
        self._celsius = valor

    @property
    def fahrenheit(self) -> float:
        return self._celsius * 9/5 + 32

    @staticmethod
    def punto_ebullicion() -> float:
        return 100.0

t = Temperatura(25)
print(f"{t.celsius}C = {t.fahrenheit}F")
t.celsius = 100
print(f"{t.celsius}C = {t.fahrenheit}F")
print("Ebullicion:", Temperatura.punto_ebullicion())
```
