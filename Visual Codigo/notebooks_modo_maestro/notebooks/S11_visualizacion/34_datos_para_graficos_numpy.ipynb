## Metodos y funciones usados en esta sesion

- np.linspace(a, b, n) — n puntos equiespaciados entre a y b (incluye ambos extremos).
- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".
- np.random.seed(n) — fija la semilla aleatoria para resultados reproducibles.
- np.random.normal(media, std, n) — n muestras de una distribucion normal.
- np.histogram(datos, bins) — cuenta cuantos valores caen en cada intervalo (bin).
- enumerate(it) — recorre un iterable devolviendo pares (indice, valor).

## Codigo de ejemplo

```python
import numpy as np

x = np.linspace(0, 2 * np.pi, 10)
y_sin = np.sin(x).round(3)
y_cos = np.cos(x).round(3)

print("x (10 puntos 0..2*pi):", x.round(2))
print("sin(x):", y_sin)
print("cos(x):", y_cos)

print(f"\nsin(0)={np.sin(0):.1f}  sin(pi/2)={np.sin(np.pi/2):.1f}  sin(pi)={np.sin(np.pi):.4f}")

np.random.seed(42)
alturas = np.random.normal(170, 10, 100)
hist, edges = np.histogram(alturas, bins=5)
print("\nHistograma alturas:")
for i, cnt in enumerate(hist):
    print(f"  {edges[i]:.0f}-{edges[i+1]:.0f}: {cnt} personas")
```
