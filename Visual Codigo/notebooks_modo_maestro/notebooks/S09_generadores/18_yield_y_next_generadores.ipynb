## Metodos y funciones usados en esta sesion

- range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.
- yield — pausa la funcion y la convierte en generador; produce un valor cada vez que se le pide con next().
- next(gen) — pide el siguiente valor de un generador/iterador.

## Codigo de ejemplo

```python
def cuadrados_lista(n):
    return [i**2 for i in range(n)]

def cuadrados_gen(n):
    for i in range(n):
        yield i**2

lst = cuadrados_lista(5)
print("Lista:", lst)

gen = cuadrados_gen(5)
print("Primero:", next(gen))
print("Segundo:", next(gen))
print("Resto:", list(gen))

def contador(inicio, fin):
    while inicio <= fin:
        yield inicio
        inicio += 1

for n in contador(1, 5):
    print(n, end=" ")
print()
```
