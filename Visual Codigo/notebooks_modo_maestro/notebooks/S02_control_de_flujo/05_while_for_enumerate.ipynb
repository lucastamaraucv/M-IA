## Metodos y funciones usados en esta sesion

- range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.
- enumerate(it) — recorre un iterable devolviendo pares (indice, valor).
- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".

## Codigo de ejemplo

```python
i = 0
while True:
    if i >= 5:
        break
    print(i, end=" ")
    i += 1
print()

for n in range(10):
    if n % 2 == 0:
        continue
    print(n, end=" ")
print()

frutas = ["manzana", "pera", "naranja"]
for i, f in enumerate(frutas):
    print(f"  [{i}] {f}")
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Asigna a `i`: `0`
- Bucle **while**: mientras `True` sea verdadero:
    - **Si** `i >= 5` entonces:
        - **Sale** del bucle (`break`).
    - **Imprime** en pantalla: `i, end=' '`.
    - Actualiza `i` (`i += 1`)
- **Imprime** en pantalla: ``.
- Bucle **for**: recorre `range(10)` guardando cada valor en `n`: — *range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.*
    - **Si** `n % 2 == 0` entonces:
        - **Salta** a la siguiente vuelta del bucle (`continue`).
    - **Imprime** en pantalla: `n, end=' '`.
- **Imprime** en pantalla: ``.
- Asigna a `frutas`: `['manzana', 'pera', 'naranja']`
- Bucle **for**: recorre `enumerate(frutas)` guardando cada valor en `(i, f)`: — *enumerate(it) — recorre un iterable devolviendo pares (indice, valor).*
    - **Imprime** en pantalla: `f' [{i}] {f}'`.
