## Metodos y funciones usados en esta sesion

- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".

## Codigo de ejemplo

```python
nota = 7.5

if nota >= 9:
    cal = "Sobresaliente"
elif nota >= 7:
    cal = "Notable"
elif nota >= 5:
    cal = "Aprobado"
else:
    cal = "Suspenso"

print(f"Nota {nota}: {cal}")

edad = 20
status = "mayor" if edad >= 18 else "menor"
print(f"Con {edad} anos eres {status} de edad")
```

## Explicacion paso a paso

- **Si** `nota >= 5` entonces:
    - Asigna a `cal`: `'Aprobado'`
- **En caso contrario:**
