## Metodos y funciones usados en esta sesion

- list.append(x) — agrega x al final de la lista, en el sitio (no devuelve nada).
- list.sort(key=None, reverse=False) — ordena la lista en el sitio.
- tupla — coleccion inmutable y ordenada; se crea con parentesis y comas.
- dict.keys() — vista con todas las claves del diccionario.

## Codigo de ejemplo

```python
nums = [3, 1, 4, 1, 5, 9]
nums.append(2)
nums.sort()
print("Lista:", nums)

coordenada = (40.41, -3.70)
print("Tupla:", coordenada)

colores = {"rojo", "azul", "rojo", "verde"}
print("Set:", colores)

persona = {"nombre": "Ana", "edad": 30}
persona["ciudad"] = "Madrid"
print("Dict:", persona)
print("Keys:", list(persona.keys()))
```
