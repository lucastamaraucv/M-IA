## Metodos y funciones usados en esta sesion

- np.zeros(shape) — crea un array lleno de ceros con la forma indicada.
- ndarray/Series.mean(axis=None) — promedio; con axis=0/1 promedia por columnas/filas.
- tupla — coleccion inmutable y ordenada; se crea con parentesis y comas.
- ndarray.astype(tipo) — convierte el array a otro dtype (ej. np.uint8).
- range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.

## Codigo de ejemplo

```python
import numpy as np

img = np.zeros((10, 10), dtype=np.uint8)
img[2:8, 2:8] = 150
img[4:6, 4:6] = 255

print("Imagen (escala de grises 0-255):")
print(img)
print("Shape:", img.shape, "| Min:", img.min(), "Max:", img.max())
print("Media:", img.mean().round(2))

binaria = (img > 100).astype(np.uint8) * 255
print("\nBinarizada (>100) fila 5:", binaria[5])

invertida = 255 - img
print("Invertida centro:", invertida[4:6, 4:6].tolist())

def blur3x3(imagen):
    r = np.zeros_like(imagen, dtype=float)
    for i in range(1, imagen.shape[0]-1):
        for j in range(1, imagen.shape[1]-1):
            r[i,j] = imagen[i-1:i+2, j-1:j+2].mean()
    return r.astype(np.uint8)

suave = blur3x3(img)
print("\nBlur 3x3 fila 5:", suave[5])
```
