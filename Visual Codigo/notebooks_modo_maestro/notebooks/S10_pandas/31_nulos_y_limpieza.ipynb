## Metodos y funciones usados en esta sesion

- pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.
- df.isnull() — mascara booleana de que celdas son nulas (NaN).
- df/Series.to_dict() — convierte a diccionario (o lista de dicts) de Python puro.
- ndarray/Series.mean(axis=None) — promedio; con axis=0/1 promedia por columnas/filas.
- df.fillna(valor) — rellena los valores nulos (NaN) con el valor/estrategia indicada.
- df.dropna() — elimina las filas (o columnas) que contienen algun nulo.

## Codigo de ejemplo

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "nombre": ["Ana", "Luis", None, "Carlos"],
    "edad":   [30, None, 35, 28],
    "puntos": [90.0, 75.0, np.nan, 92.0]
})

print("Nulos por columna:", df.isnull().sum().to_dict())
print("Total nulos:", df.isnull().sum().sum())

media_edad = df["edad"].mean()
df_limpio = df.fillna({
    "nombre": "Desconocido",
    "edad": round(media_edad, 1),
    "puntos": df["puntos"].median()
})
print("\nDespues fillna:")
print(df_limpio.to_dict("records"))
print("Filas sin nulos:", len(df.dropna()), "de", len(df))
```
