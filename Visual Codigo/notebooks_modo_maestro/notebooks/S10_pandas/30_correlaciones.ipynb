## Metodos y funciones usados en esta sesion

- np.random.seed(n) — fija la semilla aleatoria para resultados reproducibles.
- pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.
- np.random.normal(media, std, n) — n muestras de una distribucion normal.
- np.random.randint(a, b, shape) — enteros aleatorios en [a, b) con la forma dada.
- ndarray.astype(tipo) — convierte el array a otro dtype (ej. np.uint8).
- df.corr() — matriz de correlacion entre columnas numericas.

## Codigo de ejemplo

```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 50
df = pd.DataFrame({
    "altura": np.random.normal(170, 10, n),
    "edad":   np.random.randint(18, 60, n).astype(float),
    "puntos": np.random.normal(75, 15, n)
})
df["peso"] = df["altura"] * 0.4 + np.random.normal(0, 5, n)

print("Correlacion altura-peso:", round(df["altura"].corr(df["peso"]), 3))
print("Correlacion altura-edad:", round(df["altura"].corr(df["edad"]), 3))
print("Correlacion edad-puntos:", round(df["edad"].corr(df["puntos"]), 3))

corr = df.corr(numeric_only=True).round(3)
print("\nMatriz de correlacion:")
print(corr)
```
