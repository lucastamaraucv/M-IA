## Metodos y funciones usados en esta sesion

- pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.
- df.pivot_table(values, index, columns, aggfunc) — tabla resumen cruzando categorias.
- df/Series.to_dict() — convierte a diccionario (o lista de dicts) de Python puro.

## Codigo de ejemplo

```python
import pandas as pd

df = pd.DataFrame({
    "producto": ["A","B","A","B","A","B","A","B"],
    "region":   ["Norte","Norte","Sur","Sur","Este","Este","Norte","Sur"],
    "ventas":   [100, 200, 150, 80, 120, 300, 90, 110],
    "cantidad": [5, 8, 6, 3, 4, 12, 3, 5]
})

pivot = df.pivot_table(
    values="ventas",
    index="region",
    columns="producto",
    aggfunc="sum",
    fill_value=0
)
print("Pivot ventas (region x producto):")
print(pivot)
print("\nTotal por region:", pivot.sum(axis=1).to_dict())
print("Total por producto:", pivot.sum(axis=0).to_dict())
```
