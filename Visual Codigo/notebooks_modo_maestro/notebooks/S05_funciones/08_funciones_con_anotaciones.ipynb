## Metodos y funciones usados en esta sesion

- dict.items() — vista de pares (clave, valor), ideal para iterar con for k, v in d.items().
- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".

## Codigo de ejemplo

```python
def estadisticas(numeros: list) -> dict:
    if not numeros:
        return {}
    return {
        "min": min(numeros),
        "max": max(numeros),
        "media": sum(numeros) / len(numeros),
        "n": len(numeros)
    }

resultado = estadisticas([3, 1, 4, 1, 5, 9, 2, 6])
for k, v in resultado.items():
    print(f"  {k}: {v}")
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Define la funcion `estadisticas(numeros: list)`:
    - **Si** `not numeros` entonces:
        - **Devuelve** `{}`.
    - **Devuelve** `{'min': min(numeros), 'max': max(numeros), 'media': sum(numeros) / len(numeros), 'n': len…`.
- Asigna a `resultado`: `estadisticas([3, 1, 4, 1, 5, 9, 2, 6])`
- Bucle **for**: recorre `resultado.items()` guardando cada valor en `(k, v)`: — *dict.items() — vista de pares (clave, valor), ideal para iterar con for k, v in d.items().*
    - **Imprime** en pantalla: `f' {k}: {v}'`.
