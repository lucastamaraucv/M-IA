## Metodos y funciones usados en esta sesion

- sorted(it, key=None, reverse=False) — devuelve una NUEVA lista ordenada (no modifica it).
- lambda args: expr — funcion anonima de una sola expresion, comun como key= en sorted/map/filter.

## Codigo de ejemplo

```python
personas = [("Ana", 30), ("Bob", 25), ("Clara", 35)]
orden_edad = sorted(personas, key=lambda p: p[1])
print(orden_edad)

def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(6))

def make_contador(inicio=0):
    cuenta = [inicio]
    def incrementar():
        cuenta[0] += 1
        return cuenta[0]
    return incrementar

c = make_contador()
print(c(), c(), c())
```
