## Metodos y funciones usados en esta sesion

- `*args` — junta los argumentos posicionales extra de una funcion en una tupla.
- `**kwargs` — junta los argumentos nombrados extra de una funcion en un diccionario.

## Codigo de ejemplo

```python
def mostrar(*args, **kwargs):
    print("Posicionales:", args)
    print("Nombrados:", kwargs)

mostrar(1, 2, 3, nombre="Ana", edad=25)

numeros = [10, 20, 30]
print(max(*numeros))

config = {"sep": " | ", "end": "\n"}
print("a", "b", "c", **config)
```
