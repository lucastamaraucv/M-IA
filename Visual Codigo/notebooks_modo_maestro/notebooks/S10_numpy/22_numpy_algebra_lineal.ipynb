## Metodos y funciones usados en esta sesion

- np.array(lista) — crea un ndarray de NumPy a partir de una lista (o lista de listas).
- np.linalg.solve(A, b) — resuelve el sistema lineal A @ x = b.
- @ — operador de multiplicacion de matrices (equivalente a np.matmul).
- np.where(cond, si, no) — array condicional elemento a elemento (vectorizado).
- np.random.seed(n) — fija la semilla aleatoria para resultados reproducibles.
- np.random.randint(a, b, shape) — enteros aleatorios en [a, b) con la forma dada.
- ndarray/Series.mean(axis=None) — promedio; con axis=0/1 promedia por columnas/filas.

## Codigo de ejemplo

```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
b = np.array([5, 10])

x = np.linalg.solve(A, b)
print("Solucion x:", x)
print("Verificacion A@x:", A @ x)

arr = np.array([-3, -1, 0, 2, 4, -5, 7])
sin_neg = np.where(arr >= 0, arr, 0)
print("Sin negativos:", sin_neg)

np.random.seed(42)
m = np.random.randint(1, 101, (3, 3))
print("Matriz aleatoria:", m)
print("Media por fila:", m.mean(axis=1).round(2))
```
