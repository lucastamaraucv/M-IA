## Metodos y funciones usados en esta sesion

- np.array(lista) — crea un ndarray de NumPy a partir de una lista (o lista de listas).
- obj.copy() — copia superficial (shallow copy) de la lista/dict.

## Codigo de ejemplo

```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Elemento [1,2]:", arr[1, 2])
print("Fila 0:", arr[0])
print("Columna 1:", arr[:, 1])
print("Submatriz:", arr[:2, 1:])
print("Mayores de 50:", arr[arr > 50])

sub = arr[0:2, 0:2].copy()
sub[0, 0] = 999
print("Original arr[0,0]:", arr[0, 0])
```
