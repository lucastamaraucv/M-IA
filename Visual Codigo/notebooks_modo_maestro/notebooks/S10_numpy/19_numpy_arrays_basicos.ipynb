## Metodos y funciones usados en esta sesion

- np.array(lista) — crea un ndarray de NumPy a partir de una lista (o lista de listas).
- np.zeros(shape) — crea un array lleno de ceros con la forma indicada.
- np.arange(a, b, paso) — como range() pero devuelve un ndarray.
- range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.
- np.linspace(a, b, n) — n puntos equiespaciados entre a y b (incluye ambos extremos).
- np.eye(n) — matriz identidad de n x n.

## Codigo de ejemplo

```python
import numpy as np

a1 = np.array([1, 2, 3, 4, 5])
a2 = np.array([[1, 2, 3], [4, 5, 6]])
a3 = np.zeros((3, 4))
a4 = np.arange(0, 20, 2)
a5 = np.linspace(0, 1, 5)
a6 = np.eye(3)

print("1D:", a1)
print("2D shape:", a2.shape, "ndim:", a2.ndim)
print("zeros 3x4 shape:", a3.shape)
print("arange:", a4)
print("linspace:", a5)
```
