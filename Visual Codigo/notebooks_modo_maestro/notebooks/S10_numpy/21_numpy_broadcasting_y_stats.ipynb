## Metodos y funciones usados en esta sesion

- np.array(lista) — crea un ndarray de NumPy a partir de una lista (o lista de listas).
- ndarray/Series.mean(axis=None) — promedio; con axis=0/1 promedia por columnas/filas.
- ndarray/Series.std() — desviacion estandar.

## Codigo de ejemplo

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
v = np.array([10, 20, 30])
print("Broadcasting a+v:", a + v)
print("a * 2:", a * 2)

data = np.array([[4, 2, 7, 1],
                 [3, 8, 5, 9],
                 [6, 0, 4, 2]])
print("Media total:", data.mean())
print("Media columnas:", data.mean(axis=0))
print("Media filas:", data.mean(axis=1))
print("Std:", data.std().round(2))
```
