## Metodos y funciones usados en esta sesion

- map(fn, it) — aplica fn a cada elemento de it (evaluacion perezosa).
- lambda args: expr — funcion anonima de una sola expresion, comun como key= en sorted/map/filter.
- filter(fn, it) — conserva solo los elementos de it para los que fn(elem) es True.
- range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.
- functools.reduce(fn, it) — acumula it aplicando fn de a pares hasta un solo resultado.

## Codigo de ejemplo

```python
from functools import reduce

numeros = [1, 2, 3, 4, 5]
dobles = list(map(lambda x: x*2, numeros))
print("map dobles:", dobles)

pares = list(filter(lambda x: x % 2 == 0, range(1, 11)))
print("filter pares:", pares)

total = reduce(lambda x, y: x + y, numeros)
print("reduce suma:", total)

palabras = ["hola", "mundo", "python"]
mayus = list(map(str.upper, palabras))
print("map upper:", mayus)
```
