## Metodos y funciones usados en esta sesion

- collections.deque — lista doblemente enlazada; append/appendleft y pop/popleft son O(1).
- deque.popleft() — saca y devuelve el elemento del extremo izquierdo (para colas FIFO).
- list.append(x) — agrega x al final de la lista, en el sitio (no devuelve nada).
- set.add(x) — agrega x al conjunto (no hace nada si ya estaba).

## Codigo de ejemplo

```python
from collections import deque

grafo = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}

def bfs(grafo, inicio):
    visitados = []
    cola = deque([inicio])
    visto = {inicio}
    while cola:
        nodo = cola.popleft()
        visitados.append(nodo)
        for vecino in grafo[nodo]:
            if vecino not in visto:
                visto.add(vecino)
                cola.append(vecino)
    return visitados

print("BFS desde A:", bfs(grafo, "A"))
print("BFS desde C:", bfs(grafo, "C"))
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Desde `collections` importa `deque`.
- Asigna a `grafo`: `{'A': ['B', 'C'], 'B': ['A', 'D'], 'C': ['A', 'D'], 'D': ['B', 'C']}`
- Define la funcion `bfs(grafo, inicio)`:
    - Asigna a `visitados`: `[]`
    - Asigna a `cola`: `deque([inicio])` — *collections.deque — lista doblemente enlazada; append/appendleft y pop/popleft son O(1).*
    - Asigna a `visto`: `{inicio}`
    - Bucle **while**: mientras `cola` sea verdadero:
        - Asigna a `nodo`: `cola.popleft()` — *deque.popleft() — saca y devuelve el elemento del extremo izquierdo (para colas FIFO).*
        - Llama a `visitados.append(nodo)` — *list.append(x) — agrega x al final de la lista, en el sitio (no devuelve nada).*.
        - Bucle **for**: recorre `grafo[nodo]` guardando cada valor en `vecino`:
            - **Si** `vecino not in visto` entonces:
                - Llama a `visto.add(vecino)` — *set.add(x) — agrega x al conjunto (no hace nada si ya estaba).*.
                - Llama a `cola.append(vecino)` — *list.append(x) — agrega x al final de la lista, en el sitio (no devuelve nada).*.
    - **Devuelve** `visitados`.
- **Imprime** en pantalla: `'BFS desde A:', bfs(grafo, 'A')`.
