## Metodos y funciones usados en esta sesion

- collections.deque — lista doblemente enlazada; append/appendleft y pop/popleft son O(1).
- list.append(x) — agrega x al final de la lista, en el sitio (no devuelve nada).
- list.pop(i=-1) — elimina y devuelve el elemento en la posicion i (por defecto, el ultimo).
- deque.popleft() — saca y devuelve el elemento del extremo izquierdo (para colas FIFO).
- heapq.heappush(heap, x) — inserta x manteniendo la propiedad de min-heap.
- heapq.heappop(heap) — extrae y devuelve el elemento mas pequeno del heap.

## Codigo de ejemplo

```python
from collections import deque
import heapq

pila = deque()
pila.append("a")
pila.append("b")
pila.append("c")
print("Stack:", list(pila))
print("Pop:", pila.pop())

cola = deque()
cola.append("primero")
cola.append("segundo")
cola.append("tercero")
print("Queue:", list(cola))
print("Dequeue:", cola.popleft())

tareas = []
heapq.heappush(tareas, (3, "Tarea C"))
heapq.heappush(tareas, (1, "Tarea A"))
heapq.heappush(tareas, (2, "Tarea B"))
print("Siguiente:", heapq.heappop(tareas))
```
