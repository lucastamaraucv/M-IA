## Codigo de ejemplo

```python
# Practica 1 - Ejercicio 3: RLE (Run Length Encoding) iterativo
# 'dddddwwoooodddaaa' -> 'd5w2o4d3a3'
def rle(string):
    salida = string[0]
    resto = string[1:]
    contador = 1

    while resto:
        caracter = resto[0]
        if caracter == salida[-1]:
            contador += 1
        else:
            salida = salida + str(contador) + caracter
            contador = 1
        resto = resto[1:]

    salida += str(contador)
    return salida

# Test Use Case
cadena = 'dddddwwoooodddaaa'
esperado = 'd5w2o4d3a3'
resultado = rle(cadena)

print("Esperado:", esperado)
print("Obtenido:", resultado)
assert esperado == resultado
print("OK: rle funciona correctamente")
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Define la funcion `rle(string)`:
    - Asigna a `salida`: `string[0]`
    - Asigna a `resto`: `string[1:]`
    - Asigna a `contador`: `1`
    - Bucle **while**: mientras `resto` sea verdadero:
        - Asigna a `caracter`: `resto[0]`
        - **Si** `caracter == salida[-1]` entonces:
            - Actualiza `contador` (`contador += 1`)
        - **En caso contrario:**
            - Asigna a `salida`: `salida + str(contador) + caracter`
            - Asigna a `contador`: `1`
        - Asigna a `resto`: `resto[1:]`
    - Actualiza `salida` (`salida += str(contador)`)
    - **Devuelve** `salida`.
- Asigna a `cadena`: `'dddddwwoooodddaaa'`
- Asigna a `esperado`: `'d5w2o4d3a3'`
- Asigna a `resultado`: `rle(cadena)`
- **Imprime** en pantalla: `'Esperado:', esperado`.
- **Imprime** en pantalla: `'Obtenido:', resultado`.
- **Verifica** (assert) que `esperado == resultado` sea verdadero — si no, el programa se detiene aqui.
- **Imprime** en pantalla: `'OK: rle funciona correctamente'`.
