## Codigo de ejemplo

```python
# Practica 1 - Ejercicio 3 (variante recursiva de RLE)
# Observa en Frames como se apila cada llamada recursiva
def rle(s):
    if not s:
        return ''
    contador = 1
    while contador < len(s) and s[contador] == s[0]:
        contador += 1
    comprimido = s[0] + str(contador)
    return comprimido + rle(s[contador:])

cadena = 'dddddwwoooodddaaa'
resultado = rle(cadena)
print("Obtenido:", resultado)
assert resultado == 'd5w2o4d3a3'
print("OK: version recursiva tambien funciona")
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Define la funcion `rle(s)`:
    - **Si** `not s` entonces:
        - **Devuelve** `''`.
    - Asigna a `contador`: `1`
    - Bucle **while**: mientras `contador < len(s) and s[contador] == s[0]` sea verdadero:
        - Actualiza `contador` (`contador += 1`)
    - Asigna a `comprimido`: `s[0] + str(contador)`
    - **Devuelve** `comprimido + rle(s[contador:])`.
- Asigna a `cadena`: `'dddddwwoooodddaaa'`
- Asigna a `resultado`: `rle(cadena)`
- **Imprime** en pantalla: `'Obtenido:', resultado`.
- **Verifica** (assert) que `resultado == 'd5w2o4d3a3'` sea verdadero — si no, el programa se detiene aqui.
- **Imprime** en pantalla: `'OK: version recursiva tambien funciona'`.
