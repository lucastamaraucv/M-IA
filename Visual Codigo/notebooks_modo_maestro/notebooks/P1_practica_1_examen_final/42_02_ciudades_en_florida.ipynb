## Metodos y funciones usados en esta sesion

- str.strip(chars=None) — quita espacios (o los caracteres indicados) al inicio y al final.
- str.split(sep) — parte el string en una lista de trozos segun el separador.
- set.add(x) — agrega x al conjunto (no hace nada si ya estaba).

## Codigo de ejemplo

```python
# Practica 1 - Ejercicio 2: contar ciudades distintas de Florida
# Version offline (dataset embebido) en vez de descargar el CSV real,
# para poder animar el algoritmo paso a paso sin depender de internet.
# Formato original: City|State short|State full|County|City alias
datos_csv = """Miami|FL|Florida|Miami-Dade|
Orlando|FL|Florida|Orange|
Tampa|FL|Florida|Hillsborough|
Miami|FL|Florida|Miami-Dade|Miami Beach
Jacksonville|FL|Florida|Duval|
Atlanta|GA|Georgia|Fulton|
Savannah|GA|Georgia|Chatham|
Naples|FL|Florida|Collier|
Sarasota|FL|Florida|Sarasota|
Houston|TX|Texas|Harris|"""

def cities_in_florida(data):
    ciudades = set()
    for linea in data.strip().splitlines():
        columnas = linea.split("|")
        if columnas[1] == "FL":
            ciudades.add(columnas[0])
    return len(ciudades)

total = cities_in_florida(datos_csv)
print("Ciudades distintas en Florida:", total)
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Asigna a `datos_csv`: `'Miami|FL|Florida|Miami-Dade|\nOrlando|FL|Florida|Orange|\nTampa|FL|Florida|Hillsborough|…`
- Define la funcion `cities_in_florida(data)`:
    - Asigna a `ciudades`: `set()`
    - Bucle **for**: recorre `data.strip().splitlines()` guardando cada valor en `linea`: — *str.strip(chars=None) — quita espacios (o los caracteres indicados) al inicio y al final.*
        - Asigna a `columnas`: `linea.split('|')` — *str.split(sep) — parte el string en una lista de trozos segun el separador.*
        - **Si** `columnas[1] == 'FL'` entonces:
            - Llama a `ciudades.add(columnas[0])` — *set.add(x) — agrega x al conjunto (no hace nada si ya estaba).*.
    - **Devuelve** `len(ciudades)`.
- Asigna a `total`: `cities_in_florida(datos_csv)`
- **Imprime** en pantalla: `'Ciudades distintas en Florida:', total`.
