## Metodos y funciones usados en esta sesion

- range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.
- list.append(x) — agrega x al final de la lista, en el sitio (no devuelve nada).

## Codigo de ejemplo

```python
# Practica 1 - Ejercicio 1 (variante): FRange calculando primero el numero de pasos
def frange(start, end, step):
    numero_de_pasos = int((end - start) / step) + 1
    salida = []
    for i in range(numero_de_pasos):
        salida.append(round(start + float(i) * step, 2))
    return salida

output = frange(30, 36, 0.4)
print("Obtenido:", output)
print("Cantidad de elementos:", len(output))
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Define la funcion `frange(start, end, step)`:
    - Asigna a `numero_de_pasos`: `int((end - start) / step) + 1`
    - Asigna a `salida`: `[]`
    - Bucle **for**: recorre `range(numero_de_pasos)` guardando cada valor en `i`: — *range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.*
        - Llama a `salida.append(round(start + float(i) * step, 2))` — *list.append(x) — agrega x al final de la lista, en el sitio (no devuelve nada).*.
    - **Devuelve** `salida`.
