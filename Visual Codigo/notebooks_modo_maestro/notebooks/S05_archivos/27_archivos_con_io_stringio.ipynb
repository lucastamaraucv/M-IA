## Metodos y funciones usados en esta sesion

- io.StringIO() — un archivo de texto que vive en memoria, sin tocar el disco.
- enumerate(it) — recorre un iterable devolviendo pares (indice, valor).
- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".
- str.strip(chars=None) — quita espacios (o los caracteres indicados) al inicio y al final.

## Codigo de ejemplo

```python
import io

buf = io.StringIO()
buf.write("Primera linea\n")
buf.write("Segunda linea\n")
buf.write("Tercera linea\n")
contenido = buf.getvalue()
print("Escrito:", repr(contenido))

lector = io.StringIO(contenido)
for i, linea in enumerate(lector, 1):
    print(f"  Linea {i}: {linea.strip()}")
```
