## Metodos y funciones usados en esta sesion

- io.StringIO() — un archivo de texto que vive en memoria, sin tocar el disco.
- csv.writer(archivo) — escribe filas como texto CSV.
- csv.DictReader(archivo) — lee CSV y entrega cada fila como diccionario {columna: valor}.
- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".

## Codigo de ejemplo

```python
import io, csv

registros = [
    ["nombre", "edad", "ciudad"],
    ["Ana",    30,     "Madrid"],
    ["Luis",   25,     "Barcelona"],
    ["Maria",  35,     "Sevilla"]
]

buf = io.StringIO()
csv.writer(buf).writerows(registros)
csv_texto = buf.getvalue()
print("CSV generado:")
print(csv_texto)

lector = csv.DictReader(io.StringIO(csv_texto))
for fila in lector:
    print(f"  {fila['nombre']} ({fila['edad']}) - {fila['ciudad']}")
```
