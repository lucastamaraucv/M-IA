## Metodos y funciones usados en esta sesion

- slicing [::-1] — invierte una secuencia (string, lista, tupla) creando una copia.
- str.upper() — devuelve el string en mayusculas.
- str.split(sep) — parte el string en una lista de trozos segun el separador.
- str.replace(viejo, nuevo) — devuelve un string nuevo con las coincidencias reemplazadas.

## Codigo de ejemplo

```python
s = "Python para IA"
print(s[0])
print(s[-2:])
print(s[7:12])
print(s[::-1])
print(s.upper())
print(s.split(" "))
print(s.replace("Python", "Aprender"))
print(len(s))
```
