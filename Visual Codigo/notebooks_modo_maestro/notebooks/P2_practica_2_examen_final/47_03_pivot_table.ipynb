## Metodos y funciones usados en esta sesion

- df.pivot_table(values, index, columns, aggfunc) — tabla resumen cruzando categorias.
- pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.

## Codigo de ejemplo

```python
import pandas as pd

# Practica 2 - Ejercicio 3: tabla pivote con suma y media de Points
# segun cada combinacion de las categorias Height x Weight
def pivotar(df):
    return df.pivot_table(values="Points", index="Height", columns="Weight", aggfunc=["sum", "mean"])

players = pd.DataFrame({
    "Name":   ["LeBron James", "Anthony Davis", "D'Angelo Russell", "Austin Reaves", "Jarred Vanderbilt", "Jaxson Hayes"],
    "Height": ["alto", "alto", "bajo", "bajo", "alto", "alto"],
    "Weight": ["pesado", "liviano", "pesado", "pesado", "liviano", "pesado"],
    "Points": [25, 48, 12, 29, 23, 51],
})

tabla = pivotar(players)
print(tabla)
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Importa `pandas` como `pd`.
- Define la funcion `pivotar(df)`:
    - **Devuelve** `df.pivot_table(values='Points', index='Height', columns='Weight', aggfunc=['sum', 'mean'])` — *df.pivot_table(values, index, columns, aggfunc) — tabla resumen cruzando categorias.*.
- Asigna a `players`: `pd.DataFrame({'Name': ['LeBron James', 'Anthony Davis', "D'Angelo Russell", 'Austin Reave…` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Asigna a `tabla`: `pivotar(players)`
- **Imprime** en pantalla: `tabla`.
