## Metodos y funciones usados en esta sesion

- str.lower() — devuelve el string en minusculas.
- str.replace(viejo, nuevo) — devuelve un string nuevo con las coincidencias reemplazadas.
- slicing [::-1] — invierte una secuencia (string, lista, tupla) creando una copia.
- raise Excepcion(...) — lanza una excepcion manualmente.
- assert condicion — lanza AssertionError si la condicion es falsa; usado para tests rapidos.
- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".

## Codigo de ejemplo

```python
def es_palindromo(texto):
    limpio = texto.lower().replace(" ", "")
    return limpio == limpio[::-1]

def suma_lista(nums):
    return sum(nums)

def maximo(lista):
    if not lista:
        raise ValueError("Lista vacia")
    return max(lista)

assert es_palindromo("Ana") == True
assert es_palindromo("oso") == True
assert es_palindromo("Python") == False
assert es_palindromo("Anita lava la tina") == True

assert suma_lista([1, 2, 3]) == 6
assert suma_lista([]) == 0
assert suma_lista([-1, 1]) == 0

assert maximo([3, 1, 4, 1, 5]) == 5

print("Todos los tests pasaron!")
print(f"palindromo 'Ana': {es_palindromo('Ana')}")
print(f"suma [1..5]: {suma_lista([1,2,3,4,5])}")
```
