## Metodos y funciones usados en esta sesion

- generator expression — (expr for x in it) es como una list comprehension pero perezosa.
- range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.

## Codigo de ejemplo

```python
cuadrados = [x**2 for x in range(1, 21) if x % 2 != 0]
print(cuadrados)

nombres = ["Ana", "Bob", "Carlos"]
longitudes = {n: len(n) for n in nombres}
print(longitudes)

pares = {x: x**2 for x in range(1, 8) if x % 2 == 0}
print(pares)
```
