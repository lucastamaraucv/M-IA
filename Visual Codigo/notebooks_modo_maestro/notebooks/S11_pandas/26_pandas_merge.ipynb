## Metodos y funciones usados en esta sesion

- pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.
- df.merge(otro, on=...) — combina dos DataFrames como un JOIN de SQL.
- df/Series.to_dict() — convierte a diccionario (o lista de dicts) de Python puro.
- df.groupby(col) — agrupa filas por el valor de una columna para agregarlas despues.

## Codigo de ejemplo

```python
import pandas as pd

clientes = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "nombre": ["Ana", "Luis", "Maria", "Carlos"]
})

pedidos = pd.DataFrame({
    "cliente_id": [1, 1, 2, 3],
    "producto": ["Laptop", "Raton", "Monitor", "Teclado"],
    "precio": [1200, 25, 300, 80]
})

resultado = clientes.merge(pedidos, left_on="id", right_on="cliente_id", how="left")
print("Pedidos:", resultado[["nombre","producto","precio"]].to_dict("records"))

total = resultado.groupby("nombre")["precio"].sum()
print("Total por cliente:", total.to_dict())
```
