## Metodos y funciones usados en esta sesion

- pd.Series(datos, index=...) — columna etiquetada de una dimension.
- df/Series.to_dict() — convierte a diccionario (o lista de dicts) de Python puro.
- ndarray/Series.mean(axis=None) — promedio; con axis=0/1 promedia por columnas/filas.

## Codigo de ejemplo

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, 50], index=["a", "b", "c", "d", "e"])
print("Serie:", s.to_dict())
print("Elemento c:", s["c"])
print("Slicing a:c:", s["a":"c"].to_dict())
print("Mayores 25:", s[s > 25].to_dict())
print("Media:", s.mean(), "Suma:", s.sum())
```
