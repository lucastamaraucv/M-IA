## Metodos y funciones usados en esta sesion

- pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.
- df.iloc[pos] — selecciona filas/columnas por posicion numerica.
- df/Series.to_dict() — convierte a diccionario (o lista de dicts) de Python puro.

## Codigo de ejemplo

```python
import pandas as pd

df = pd.DataFrame({
    "Nombre": ["Ana", "Luis", "Maria", "Carlos"],
    "Edad": [30, 25, 35, 28],
    "Ciudad": ["Madrid", "Barcelona", "Sevilla", "Valencia"]
})

print("Shape:", df.shape)
print("Nombres:", df["Nombre"].tolist())
print("Edades:", df["Edad"].tolist())
print("Mayor de 27:", df[df["Edad"] > 27]["Nombre"].tolist())
print("Iloc[0]:", df.iloc[0].to_dict())
```
