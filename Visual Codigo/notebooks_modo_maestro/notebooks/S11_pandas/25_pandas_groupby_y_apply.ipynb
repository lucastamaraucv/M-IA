## Metodos y funciones usados en esta sesion

- pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.
- Series/DataFrame.apply(fn) — aplica una funcion a cada elemento, fila o columna.
- lambda args: expr — funcion anonima de una sola expresion, comun como key= en sorted/map/filter.
- df/Series.to_dict() — convierte a diccionario (o lista de dicts) de Python puro.
- df.groupby(col) — agrupa filas por el valor de una columna para agregarlas despues.
- ndarray/Series.mean(axis=None) — promedio; con axis=0/1 promedia por columnas/filas.
- df.loc[etiqueta] — selecciona filas/columnas por etiqueta (nombre de indice).

## Codigo de ejemplo

```python
import pandas as pd

df = pd.DataFrame({
    "genero": ["F", "M", "F", "M", "F"],
    "edad":   [30, 25, 35, 28, 22],
    "puntos": [90, 75, 85, 92, 88]
})

rango = df[["edad", "puntos"]].apply(lambda c: c.max() - c.min())
print("Rango por columna:", rango.to_dict())

por_genero = df.groupby("genero")[["edad","puntos"]].mean()
print("Media F:", por_genero.loc["F"].to_dict())
print("Media M:", por_genero.loc["M"].to_dict())
```
