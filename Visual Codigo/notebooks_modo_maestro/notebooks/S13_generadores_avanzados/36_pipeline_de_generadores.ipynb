## Metodos y funciones usados en esta sesion

- yield — pausa la funcion y la convierte en generador; produce un valor cada vez que se le pide con next().

## Codigo de ejemplo

```python
def leer(datos):
    for item in datos:
        yield item

def filtrar(stream, umbral):
    for item in stream:
        if item > umbral:
            yield item

def transformar(stream, factor):
    for item in stream:
        yield item * factor

def acumular(stream):
    total = 0
    for item in stream:
        total += item
        yield total

datos = [1, 5, 3, 8, 2, 9, 4, 7, 6]
pipeline = list(acumular(transformar(filtrar(leer(datos), 4), 2)))
print("Datos:", datos)
print("Filtrar >4, x2, acumular:", pipeline)

total_pipeline = sum(transformar(filtrar(leer(datos), 4), 2))
print("Suma total del pipeline:", total_pipeline)
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Define la funcion `leer(datos)`:
    - Bucle **for**: recorre `datos` guardando cada valor en `item`:
        - **yield** `item` — produce un valor y pausa la funcion hasta el proximo `next()`.
- Define la funcion `filtrar(stream, umbral)`:
    - Bucle **for**: recorre `stream` guardando cada valor en `item`:
        - **Si** `item > umbral` entonces:
            - **yield** `item` — produce un valor y pausa la funcion hasta el proximo `next()`.
- Define la funcion `transformar(stream, factor)`:
    - Bucle **for**: recorre `stream` guardando cada valor en `item`:
        - **yield** `item * factor` — produce un valor y pausa la funcion hasta el proximo `next()`.
- Define la funcion `acumular(stream)`:
    - Asigna a `total`: `0`
    - Bucle **for**: recorre `stream` guardando cada valor en `item`:
        - Actualiza `total` (`total += item`)
        - **yield** `total` — produce un valor y pausa la funcion hasta el proximo `next()`.
- Asigna a `datos`: `[1, 5, 3, 8, 2, 9, 4, 7, 6]`
- Asigna a `pipeline`: `list(acumular(transformar(filtrar(leer(datos), 4), 2)))`
