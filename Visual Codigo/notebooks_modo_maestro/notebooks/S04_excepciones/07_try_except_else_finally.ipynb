## Metodos y funciones usados en esta sesion

- try/except/else/finally — captura errores en tiempo de ejecucion sin romper el programa.
- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".

## Codigo de ejemplo

```python
def dividir(a, b):
    try:
        resultado = a / b
    except ZeroDivisionError:
        print("Error: division por cero")
        return None
    else:
        print(f"{a} / {b} = {resultado}")
        return resultado
    finally:
        print("--- fin operacion ---")

dividir(10, 2)
dividir(5, 0)
```

## Explicacion paso a paso

Que hace cada bloque del codigo de arriba, en orden de ejecucion:

- Define la funcion `dividir(a, b)`:
    - Intenta ejecutar (**try**):
        - Asigna a `resultado`: `a / b`
    - Si ocurre `ZeroDivisionError` (**except**):
        - **Imprime** en pantalla: `'Error: division por cero'`.
        - **Devuelve** `None`.
    - Si no hubo ninguna excepcion (**else**):
        - **Imprime** en pantalla: `f'{a} / {b} = {resultado}'`.
        - **Devuelve** `resultado`.
    - Siempre al final, haya habido error o no (**finally**):
        - **Imprime** en pantalla: `'--- fin operacion ---'`.
- Llama a `dividir(10, 2)`.
- Llama a `dividir(5, 0)`.
