## Metodos y funciones usados en esta sesion

- zip(a, b) — combina dos (o mas) iterables en pares/tuplas, elemento a elemento.
- list.sort(key=None, reverse=False) — ordena la lista en el sitio.
- lambda args: expr — funcion anonima de una sola expresion, comun como key= en sorted/map/filter.
- collections.Counter — diccionario especializado para contar apariciones de elementos.
- Counter.most_common(n) — los n elementos mas frecuentes, de mayor a menor.
- f-string — interpolacion de expresiones dentro de un string con f"...{expr}...".

## Codigo de ejemplo

```python
import math
from collections import Counter

def distancia(p1, p2):
    return math.sqrt(sum((a-b)**2 for a,b in zip(p1,p2)))

def knn_predecir(X_train, y_train, punto, k=3):
    dists = [(distancia(punto, x), y) for x, y in zip(X_train, y_train)]
    dists.sort(key=lambda t: t[0])
    k_vecinos = [y for _, y in dists[:k]]
    return Counter(k_vecinos).most_common(1)[0][0]

# 2 clases: grupo A (cerca de [1,1]) y B (cerca de [7,7])
X_train = [[1,1],[1.5,2],[2,1.5],[2.5,2],
           [7,7],[7.5,8],[8,7.5],[8.5,8]]
y_train = [  0,    0,     0,      0,
             1,    1,     1,      1]

X_test  = [[1.8, 1.9], [7.8, 7.5], [2.1, 2.2], [6.9, 7.1]]
y_real  = [0, 1, 0, 1]

print("Predicciones KNN (k=3):")
correctas = 0
for punto, real in zip(X_test, y_real):
    pred = knn_predecir(X_train, y_train, punto)
    ok = pred == real
    if ok: correctas += 1
    print(f"  {punto} -> pred={pred} real={real} {'OK' if ok else 'X'}")

print(f"Precision: {correctas}/{len(y_real)} = {correctas/len(X_test):.1%}")
print("(sklearn hace esto con: KNeighborsClassifier(n_neighbors=3).fit(X, y))")
```
