## Grafico de linea basico

plt.plot dibuja una serie de puntos (x, y) conectados por una linea.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Funcion seno")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `matplotlib.pyplot` como `plt`.
- Importa `numpy` como `np`.
- Asigna a `x`: `np.linspace(0, 10, 100)` — *np.linspace(a, b, n) — n puntos equiespaciados entre a y b (incluye ambos extremos).*
- Asigna a `y`: `np.sin(x)`
- Llama a `plt.plot(x, y)`.
- Llama a `plt.title('Funcion seno')` — *str.title() — pone en mayuscula la primera letra de cada palabra.*.
- Llama a `plt.xlabel('x')`.
- Llama a `plt.ylabel('sin(x)')`.
- Llama a `plt.show()`.

## Subplots: varias graficas en una figura

plt.subplots(filas, columnas) devuelve una figura y un array de ejes.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
fig, ejes = plt.subplots(1, 2, figsize=(10, 4))

ejes[0].plot(x, np.sin(x), color="tab:blue")
ejes[0].set_title("Seno")

ejes[1].plot(x, np.cos(x), color="tab:orange")
ejes[1].set_title("Coseno")

plt.tight_layout()
plt.show()
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `matplotlib.pyplot` como `plt`.
- Importa `numpy` como `np`.
- Asigna a `x`: `np.linspace(0, 10, 100)` — *np.linspace(a, b, n) — n puntos equiespaciados entre a y b (incluye ambos extremos).*
- Asigna a `(fig, ejes)`: `plt.subplots(1, 2, figsize=(10, 4))`
- Llama a `ejes[0].plot(x, np.sin(x), color='tab:blue')`.
- Llama a `ejes[0].set_title('Seno')`.
- Llama a `ejes[1].plot(x, np.cos(x), color='tab:orange')`.
- Llama a `ejes[1].set_title('Coseno')`.
- Llama a `plt.tight_layout()`.
- Llama a `plt.show()`.

## Grafico de barras

```python
import matplotlib.pyplot as plt

productos = ["A", "B", "C", "D"]
ventas = [120, 85, 200, 60]

plt.bar(productos, ventas, color="tab:green")
plt.title("Ventas por producto")
plt.ylabel("Unidades vendidas")
plt.show()
```

## Scatter (dispersion)

plt.scatter muestra cada punto por separado — ideal para ver correlacion entre dos variables.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(0)
x = rng.normal(0, 1, 100)
y = x * 2 + rng.normal(0, 1, 100)

plt.scatter(x, y, alpha=0.6, c=y, cmap="viridis")
plt.title("Dispersion x vs y")
plt.colorbar(label="valor de y")
plt.show()
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `matplotlib.pyplot` como `plt`.
- Importa `numpy` como `np`.
- Asigna a `rng`: `np.random.default_rng(0)`
- Asigna a `x`: `rng.normal(0, 1, 100)`
- Asigna a `y`: `x * 2 + rng.normal(0, 1, 100)`
- Llama a `plt.scatter(x, y, alpha=0.6, c=y, cmap='viridis')`.
- Llama a `plt.title('Dispersion x vs y')` — *str.title() — pone en mayuscula la primera letra de cada palabra.*.
- Llama a `plt.colorbar(label='valor de y')`.
- Llama a `plt.show()`.

## Histograma

plt.hist agrupa valores en bins para mostrar como se distribuyen.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(0)
datos = rng.normal(170, 10, 500)

plt.hist(datos, bins=20, color="tab:purple", edgecolor="white")
plt.title("Distribucion de alturas")
plt.xlabel("cm")
plt.ylabel("frecuencia")
plt.show()
```

## Personalizacion: colores, estilos y leyenda

Lineas punteadas, colores por nombre, `label=` + `plt.legend()`.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 50)

plt.plot(x, np.sin(x), "--", color="crimson", label="seno")
plt.plot(x, np.cos(x), ":", color="navy", label="coseno")
plt.legend()
plt.grid(True, alpha=0.3)
plt.title("Seno y coseno")
plt.show()
```

## Estilo orientado a objetos (fig, ax)

La forma recomendada para graficos complejos: controlar la figura y los ejes directamente.

```python
import matplotlib.pyplot as plt
import numpy as np
import io

fig, ax = plt.subplots(figsize=(6, 4))
x = np.linspace(-5, 5, 100)
ax.plot(x, x**2)
ax.set_title("Parabola")
ax.set_xlabel("x")
ax.set_ylabel("x^2")
ax.axhline(0, color="gray", linewidth=0.5)
ax.axvline(0, color="gray", linewidth=0.5)

buffer = io.BytesIO()
fig.savefig(buffer, format="png", dpi=100)
print("Figura serializada en memoria:", len(buffer.getvalue()), "bytes")
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `matplotlib.pyplot` como `plt`.
- Importa `numpy` como `np`.
- Importa `io`.
- Asigna a `(fig, ax)`: `plt.subplots(figsize=(6, 4))`
- Asigna a `x`: `np.linspace(-5, 5, 100)` — *np.linspace(a, b, n) — n puntos equiespaciados entre a y b (incluye ambos extremos).*
- Llama a `ax.plot(x, x ** 2)`.
- Llama a `ax.set_title('Parabola')`.
- Llama a `ax.set_xlabel('x')`.
- Llama a `ax.set_ylabel('x^2')`.
- Llama a `ax.axhline(0, color='gray', linewidth=0.5)`.
- Llama a `ax.axvline(0, color='gray', linewidth=0.5)`.
- Asigna a `buffer`: `io.BytesIO()`
- Llama a `fig.savefig(buffer, format='png', dpi=100)`.
- **Imprime** en pantalla: `'Figura serializada en memoria:', len(buffer.getvalue()), 'bytes'`.

## Anotaciones y texto en el grafico

plt.annotate senala un punto con texto y una flecha.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)
maximo_idx = np.argmax(y)

plt.plot(x, y)
plt.annotate("maximo", xy=(x[maximo_idx], y[maximo_idx]),
             xytext=(x[maximo_idx] + 1, y[maximo_idx] - 0.3),
             arrowprops=dict(arrowstyle="->"))
plt.title("Seno con anotacion")
plt.show()
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `matplotlib.pyplot` como `plt`.
- Importa `numpy` como `np`.
- Asigna a `x`: `np.linspace(0, 10, 100)` — *np.linspace(a, b, n) — n puntos equiespaciados entre a y b (incluye ambos extremos).*
- Asigna a `y`: `np.sin(x)`
- Asigna a `maximo_idx`: `np.argmax(y)`
- Llama a `plt.plot(x, y)`.
- Llama a `plt.annotate('maximo', xy=(x[maximo_idx], y[maximo_idx]), xytext=(x[maximo_idx] + 1, y[maximo_idx] - 0.3), arrowprops=dict(arrowstyle='->'))`.
- Llama a `plt.title('Seno con anotacion')` — *str.title() — pone en mayuscula la primera letra de cada palabra.*.
- Llama a `plt.show()`.
