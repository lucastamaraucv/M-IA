# Verificar el entorno: ¿tengo todas las librerías del Modo Maestro?

Corre la celda de abajo en tu Jupyter (Anaconda). Comprueba, importando de
verdad cada módulo, que estén disponibles:

- **Estándar de Python** (`abc`, `collections`, `csv`, `functools`, `heapq`,
  `io`, `math`, `string`, `sys`) — vienen con cualquier instalación de
  Python, nunca hace falta instalarlas aparte. Son las que usan las 48
  sesiones del Modo Maestro además de `numpy`/`pandas`.
- **De terceros** (`numpy`, `pandas` — usadas en las sesiones; `matplotlib`,
  `seaborn` — usadas en esta carpeta `referencia/`) — necesitan estar
  instaladas, pero **vienen preinstaladas por defecto en Anaconda**.

Si algo falta, la celda te da el comando exacto para instalarlo.

```python
import importlib
import sys

STDLIB = ["abc", "collections", "csv", "functools", "heapq", "io", "math", "string", "sys"]
THIRD_PARTY = ["numpy", "pandas", "matplotlib", "seaborn"]

print("Python:", sys.version.split()[0])
print("Ejecutable:", sys.executable)
print()

print("== Bibliotecas estandar (siempre disponibles) ==")
for mod in STDLIB:
    importlib.import_module(mod)
    print(f"  OK    {mod}")

print()
print("== Librerias de terceros ==")
faltantes = []
for mod in THIRD_PARTY:
    try:
        m = importlib.import_module(mod)
        v = getattr(m, "__version__", "?")
        print(f"  OK    {mod:10s} v{v}")
    except ImportError:
        faltantes.append(mod)
        print(f"  FALTA {mod:10s} -> instalar con: conda install {mod}   (o pip install {mod})")

print()
if faltantes:
    print(f"Te faltan {len(faltantes)} librerias: {', '.join(faltantes)}")
else:
    print("Todo listo: podes correr las 48 sesiones del Modo Maestro y los notebooks de referencia/ sin instalar nada mas.")
```
