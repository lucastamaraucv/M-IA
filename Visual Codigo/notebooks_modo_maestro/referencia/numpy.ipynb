## Creacion de arrays

## Indexado, slicing y mascaras booleanas

Acceder a sub-arrays y filtrar por condicion sin bucles.

## Broadcasting

```python
import numpy as np

matriz = np.array([[1, 2, 3], [4, 5, 6]])
vector = np.array([10, 20, 30])
print(matriz + vector)

columna = np.array([[1], [2]])
print(matriz + columna)

print(matriz * 2)
```

## Operaciones vectorizadas vs bucles Python

Por que se prefiere `array * 2` a un `for` — mismo resultado, mucho mas rapido.

```python
import numpy as np
import time

n = 200_000
datos = np.arange(n)

inicio = time.perf_counter()
resultado_vectorizado = datos * 2 + 1
tiempo_vector = time.perf_counter() - inicio

inicio = time.perf_counter()
resultado_bucle = [x * 2 + 1 for x in range(n)]
tiempo_bucle = time.perf_counter() - inicio

print(f"Vectorizado: {tiempo_vector:.5f}s")
print(f"Bucle Python: {tiempo_bucle:.5f}s")
print("Mismo resultado:", list(resultado_vectorizado[:5]) == resultado_bucle[:5])
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `numpy` como `np`.
- Importa `time`.
- Asigna a `n`: `200000`
- Asigna a `datos`: `np.arange(n)` — *np.arange(a, b, paso) — como range() pero devuelve un ndarray.*
- Asigna a `inicio`: `time.perf_counter()`
- Asigna a `resultado_vectorizado`: `datos * 2 + 1`
- Asigna a `tiempo_vector`: `time.perf_counter() - inicio`
- Asigna a `inicio`: `time.perf_counter()`
- Asigna a `resultado_bucle`: `[x * 2 + 1 for x in range(n)]` — *generator expression — (expr for x in it) es como una list comprehension pero perezosa.*
- Asigna a `tiempo_bucle`: `time.perf_counter() - inicio`
- **Imprime** en pantalla: `f'Vectorizado: {tiempo_vector:.5f}s'`.
- **Imprime** en pantalla: `f'Bucle Python: {tiempo_bucle:.5f}s'`.
- **Imprime** en pantalla: `'Mismo resultado:', list(resultado_vectorizado[:5]) == resultado_bucle[:5]`.

## Algebra lineal

Resolver sistemas de ecuaciones y operar con matrices.

```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
b = np.array([5, 10])

x = np.linalg.solve(A, b)
print("Solucion:", x)
print("Verificacion A@x:", A @ x)
print("Transpuesta:\n", A.T)
print("Inversa:\n", np.linalg.inv(A))
print("Determinante:", np.linalg.det(A))
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `numpy` como `np`.
- Asigna a `A`: `np.array([[2, 1], [1, 3]])` — *np.array(lista) — crea un ndarray de NumPy a partir de una lista (o lista de listas).*
- Asigna a `b`: `np.array([5, 10])` — *np.array(lista) — crea un ndarray de NumPy a partir de una lista (o lista de listas).*
- Asigna a `x`: `np.linalg.solve(A, b)` — *np.linalg.solve(A, b) — resuelve el sistema lineal A @ x = b.*
- **Imprime** en pantalla: `'Solucion:', x`.
- **Imprime** en pantalla: `'Verificacion A@x:', A @ x`.
- **Imprime** en pantalla: `'Transpuesta:\n', A.T`.
- **Imprime** en pantalla: `'Inversa:\n', np.linalg.inv(A)`.
- **Imprime** en pantalla: `'Determinante:', np.linalg.det(A)`.

## Ordenar y buscar

sort, argsort, unique y where — sin escribir el algoritmo a mano.

```python
import numpy as np

datos = np.array([5, 2, 9, 1, 5, 6])
print("Ordenado:", np.sort(datos))
print("Indices que ordenan:", np.argsort(datos))
print("Unicos:", np.unique(datos))
print("Donde es > 4:", np.where(datos > 4))
print("Reemplazo condicional:", np.where(datos > 4, datos, 0))
```

## Numeros aleatorios reproducibles

`default_rng(seed)` — la forma moderna de generar aleatoriedad reproducible.

```python
import numpy as np

rng = np.random.default_rng(seed=42)
print("Enteros:", rng.integers(1, 100, size=5))
print("Normal:", rng.normal(loc=0, scale=1, size=5).round(3))
print("Eleccion:", rng.choice(["a", "b", "c", "d"], size=3))
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `numpy` como `np`.
- Asigna a `rng`: `np.random.default_rng(seed=42)`
- **Imprime** en pantalla: `'Enteros:', rng.integers(1, 100, size=5)`.
- **Imprime** en pantalla: `'Normal:', rng.normal(loc=0, scale=1, size=5).round(3)`.
- **Imprime** en pantalla: `'Eleccion:', rng.choice(['a', 'b', 'c', 'd'], size=3)`.
