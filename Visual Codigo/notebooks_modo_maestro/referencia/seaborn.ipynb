## Preparar datos y tema (sns.set_theme)

sns.set_theme() cambia el aspecto de TODOS los graficos siguientes, sean de seaborn o matplotlib.

```python
import seaborn as sns
import pandas as pd
import numpy as np

sns.set_theme(style="whitegrid")

rng = np.random.default_rng(7)
df = pd.DataFrame({
    "horas_estudio": rng.uniform(0, 10, 80),
    "grupo": rng.choice(["A", "B"], 80),
})
df["nota"] = 4 + df["horas_estudio"] * 0.6 + rng.normal(0, 1, 80)
print(df.head())
```

## scatterplot con hue

`hue=` colorea los puntos segun una columna categorica, para comparar grupos en el mismo grafico.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

rng = np.random.default_rng(7)
df = pd.DataFrame({
    "horas_estudio": rng.uniform(0, 10, 80),
    "grupo": rng.choice(["A", "B"], 80),
})
df["nota"] = 4 + df["horas_estudio"] * 0.6 + rng.normal(0, 1, 80)

sns.scatterplot(data=df, x="horas_estudio", y="nota", hue="grupo")
plt.title("Nota segun horas de estudio")
plt.show()
```

## histplot: distribucion de una variable

Histograma + estimador de densidad (`kde=True`) en una sola llamada.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(1)
alturas = rng.normal(170, 8, 300)

sns.histplot(alturas, bins=25, kde=True, color="teal")
plt.title("Distribucion de alturas (con estimador de densidad)")
plt.show()
```

## boxplot y violinplot: comparar grupos

## heatmap: matriz de correlacion

Combina `df.corr()` de Pandas con `sns.heatmap` para ver relaciones entre variables de un vistazo.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

rng = np.random.default_rng(5)
df = pd.DataFrame({
    "altura": rng.normal(170, 10, 100),
    "peso": rng.normal(70, 12, 100),
    "edad": rng.integers(18, 60, 100),
})
df["peso"] = df["peso"] + df["altura"] * 0.3

sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlacion entre variables")
plt.show()
```

- Asigna a `df`: `pd.DataFrame({'altura': rng.normal(170, 10, 100), 'peso': rng.normal(70, 12, 100), 'edad'…` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Asigna a `df['peso']`: `df['peso'] + df['altura'] * 0.3`

## barplot con intervalo de confianza

Cada barra muestra el promedio; la lineita vertical es la variabilidad (`errorbar=`).

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

rng = np.random.default_rng(9)
df = pd.DataFrame({
    "producto": rng.choice(["A", "B", "C"], 120),
    "satisfaccion": rng.uniform(1, 5, 120),
})

sns.barplot(data=df, x="producto", y="satisfaccion", errorbar="sd")
plt.title("Satisfaccion promedio por producto (barra = desvio estandar)")
plt.show()
```

## lineplot con intervalo de confianza (series temporales)

Con datos repetidos por `x`, seaborn dibuja el promedio y una banda de variabilidad automaticamente.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

rng = np.random.default_rng(13)
dias = list(range(30)) * 3
valores = []
for repeticion in range(3):
    valores += list(20 + np.cumsum(rng.normal(0, 1, 30)))

df = pd.DataFrame({"dia": dias, "valor": valores})

sns.lineplot(data=df, x="dia", y="valor", errorbar="sd")
plt.title("Evolucion diaria (con banda de variabilidad)")
plt.show()
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `seaborn` como `sns`.
- Importa `matplotlib.pyplot` como `plt`.
- Importa `pandas` como `pd`.
- Importa `numpy` como `np`.
- Asigna a `rng`: `np.random.default_rng(13)`
- Asigna a `dias`: `list(range(30)) * 3` — *range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.*
- Asigna a `valores`: `[]`
- Bucle **for**: recorre `range(3)` guardando cada valor en `repeticion`: — *range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.*
    - Actualiza `valores` (`valores += list(20 + np.cumsum(rng.normal(0, 1, 30)))`)
- Asigna a `df`: `pd.DataFrame({'dia': dias, 'valor': valores})` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Llama a `sns.lineplot(data=df, x='dia', y='valor', errorbar='sd')`.
- Llama a `plt.title('Evolucion diaria (con banda de variabilidad)')` — *str.title() — pone en mayuscula la primera letra de cada palabra.*.
- Llama a `plt.show()`.
