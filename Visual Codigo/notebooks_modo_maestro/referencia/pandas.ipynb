## Seleccion: loc, iloc y filtrado booleano

loc selecciona por etiqueta, iloc por posicion; los boolean masks filtran filas.

```python
import pandas as pd

df = pd.DataFrame({
    "producto": ["Laptop", "Mouse", "Teclado", "Monitor"],
    "precio": [1200, 25, 80, 300],
    "stock": [5, 50, 30, 12],
})

print(df.loc[0, "producto"])
print(df.iloc[1:3])
print(df[df["precio"] > 100])
print(df.loc[df["stock"] < 20, ["producto", "stock"]])
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `pandas` como `pd`.
- Asigna a `df`: `pd.DataFrame({'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor'], 'precio': [1200, 25,…` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- **Imprime** en pantalla: `df.loc[0, 'producto']`.
- **Imprime** en pantalla: `df.iloc[1:3]`.
- **Imprime** en pantalla: `df[df['precio'] > 100]`.
- **Imprime** en pantalla: `df.loc[df['stock'] < 20, ['producto', 'stock']]`.

## Ordenar y rankear

sort_values ordena filas; rank asigna una posicion relativa a cada valor.

```python
import pandas as pd

df = pd.DataFrame({
    "producto": ["Laptop", "Mouse", "Teclado", "Monitor"],
    "precio": [1200, 25, 80, 300],
})

print(df.sort_values("precio", ascending=False))
print(df.sort_values("precio").reset_index(drop=True))
df["ranking"] = df["precio"].rank(ascending=False).astype(int)
print(df)
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `pandas` como `pd`.
- Asigna a `df`: `pd.DataFrame({'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor'], 'precio': [1200, 25,…` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- **Imprime** en pantalla: `df.sort_values('precio', ascending=False)`.
- **Imprime** en pantalla: `df.sort_values('precio').reset_index(drop=True)`.
- Asigna a `df['ranking']`: `df['precio'].rank(ascending=False).astype(int)` — *ndarray.astype(tipo) — convierte el array a otro dtype (ej. np.uint8).*
- **Imprime** en pantalla: `df`.

## groupby y agregaciones multiples

Agrupar filas por una o varias columnas y resumirlas con `.agg([...])`.

```python
import pandas as pd

ventas = pd.DataFrame({
    "region": ["Norte", "Sur", "Norte", "Sur", "Norte"],
    "producto": ["A", "A", "B", "B", "A"],
    "monto": [100, 150, 200, 90, 120],
})

por_region = ventas.groupby("region")["monto"].agg(["sum", "mean", "count"])
print(por_region)

por_ambos = ventas.groupby(["region", "producto"])["monto"].sum()
print(por_ambos)
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `pandas` como `pd`.
- Asigna a `ventas`: `pd.DataFrame({'region': ['Norte', 'Sur', 'Norte', 'Sur', 'Norte'], 'producto': ['A', 'A',…` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Asigna a `por_region`: `ventas.groupby('region')['monto'].agg(['sum', 'mean', 'count'])` — *df.groupby(col) — agrupa filas por el valor de una columna para agregarlas despues.*
- **Imprime** en pantalla: `por_region`.
- Asigna a `por_ambos`: `ventas.groupby(['region', 'producto'])['monto'].sum()` — *df.groupby(col) — agrupa filas por el valor de una columna para agregarlas despues.*
- **Imprime** en pantalla: `por_ambos`.

## merge, join y concat

Combinar tablas: merge cruza por clave (como un JOIN de SQL), concat apila filas.

```python
import pandas as pd

clientes = pd.DataFrame({"id": [1, 2, 3], "nombre": ["Ana", "Luis", "Eva"]})
pedidos = pd.DataFrame({"cliente_id": [1, 1, 2], "monto": [100, 50, 200]})

combinado = clientes.merge(pedidos, left_on="id", right_on="cliente_id", how="left")
print(combinado)

extra = pd.DataFrame({"id": [4], "nombre": ["Marcos"]})
todos = pd.concat([clientes, extra], ignore_index=True)
print(todos)
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `pandas` como `pd`.
- Asigna a `clientes`: `pd.DataFrame({'id': [1, 2, 3], 'nombre': ['Ana', 'Luis', 'Eva']})` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Asigna a `pedidos`: `pd.DataFrame({'cliente_id': [1, 1, 2], 'monto': [100, 50, 200]})` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Asigna a `combinado`: `clientes.merge(pedidos, left_on='id', right_on='cliente_id', how='left')` — *df.merge(otro, on=...) — combina dos DataFrames como un JOIN de SQL.*
- **Imprime** en pantalla: `combinado`.
- Asigna a `extra`: `pd.DataFrame({'id': [4], 'nombre': ['Marcos']})` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Asigna a `todos`: `pd.concat([clientes, extra], ignore_index=True)`
- **Imprime** en pantalla: `todos`.

## pivot_table y crosstab

Tablas resumen cruzando dos categorias, con una funcion de agregacion.

```python
import pandas as pd

ventas = pd.DataFrame({
    "region": ["Norte", "Sur", "Norte", "Sur"],
    "producto": ["A", "A", "B", "B"],
    "monto": [100, 150, 200, 90],
})

pivote = ventas.pivot_table(values="monto", index="region", columns="producto", aggfunc="sum", fill_value=0)
print(pivote)

tabla_cruzada = pd.crosstab(ventas["region"], ventas["producto"])
print(tabla_cruzada)
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `pandas` como `pd`.

- Asigna a `pivote`: `ventas.pivot_table(values='monto', index='region', columns='producto', aggfunc='sum', fil…` — *df.pivot_table(values, index, columns, aggfunc) — tabla resumen cruzando categorias.*
- **Imprime** en pantalla: `pivote`.
- Asigna a `tabla_cruzada`: `pd.crosstab(ventas['region'], ventas['producto'])`
- **Imprime** en pantalla: `tabla_cruzada`.

## Datos faltantes

isna/fillna/dropna — detectar, rellenar o eliminar valores nulos (NaN).

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "nombre": ["Ana", "Luis", None, "Eva"],
    "edad": [30, np.nan, 25, 40],
})

print(df.isna().sum())
print(df.fillna({"nombre": "Desconocido", "edad": df["edad"].mean()}))
print(df.dropna())
```

- Asigna a `df`: `pd.DataFrame({'nombre': ['Ana', 'Luis', None, 'Eva'], 'edad': [30, np.nan, 25, 40]})` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*

## apply, map y metodos de texto vectorizados

Transformar columnas con una funcion propia, o con el accessor `.str`.

```python
import pandas as pd

df = pd.DataFrame({
    "nombre": [" ana garcia ", "LUIS PEREZ", "eva  torres"],
    "puntos": [10, 25, 18],
})

df["nombre_limpio"] = df["nombre"].str.strip().str.title()
df["categoria"] = df["puntos"].map(lambda p: "alto" if p >= 20 else "bajo")
df["doble_puntos"] = df["puntos"].apply(lambda p: p * 2)
print(df)
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `pandas` como `pd`.
- Asigna a `df`: `pd.DataFrame({'nombre': [' ana garcia ', 'LUIS PEREZ', 'eva torres'], 'puntos': [10, 25, …` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Asigna a `df['nombre_limpio']`: `df['nombre'].str.strip().str.title()` — *str.strip(chars=None) — quita espacios (o los caracteres indicados) al inicio y al final.*
- Asigna a `df['categoria']`: `df['puntos'].map(lambda p: 'alto' if p >= 20 else 'bajo')` — *lambda args: expr — funcion anonima de una sola expresion, comun como key= en sorted/map/filter.*
- Asigna a `df['doble_puntos']`: `df['puntos'].apply(lambda p: p * 2)` — *Series/DataFrame.apply(fn) — aplica una funcion a cada elemento, fila o columna.*
- **Imprime** en pantalla: `df`.

## Fechas con to_datetime

Convertir texto a fechas reales y extraer componentes (mes, dia de la semana...).

```python
import pandas as pd

fechas = pd.to_datetime(["2026-01-15", "2026-03-02", "2026-07-30"])
df = pd.DataFrame({"fecha": fechas, "ventas": [100, 150, 200]})
df["mes"] = df["fecha"].dt.month
df["dia_semana"] = df["fecha"].dt.day_name()
print(df)
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `pandas` como `pd`.
- Asigna a `fechas`: `pd.to_datetime(['2026-01-15', '2026-03-02', '2026-07-30'])`
- Asigna a `df`: `pd.DataFrame({'fecha': fechas, 'ventas': [100, 150, 200]})` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Asigna a `df['mes']`: `df['fecha'].dt.month`
- Asigna a `df['dia_semana']`: `df['fecha'].dt.day_name()`
- **Imprime** en pantalla: `df`.

## Leer y escribir CSV en memoria

to_csv/read_csv funcionan igual con un archivo real o con io.StringIO.

```python
import pandas as pd
import io

df = pd.DataFrame({"a": [1, 2, 3], "b": ["x", "y", "z"]})
buffer = io.StringIO()
df.to_csv(buffer, index=False)
print(buffer.getvalue())

buffer.seek(0)
leido = pd.read_csv(buffer)
print(leido)
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `pandas` como `pd`.
- Importa `io`.
- Asigna a `df`: `pd.DataFrame({'a': [1, 2, 3], 'b': ['x', 'y', 'z']})` — *pd.DataFrame(dict) — tabla 2D; cada clave del dict se vuelve una columna.*
- Asigna a `buffer`: `io.StringIO()` — *io.StringIO() — un archivo de texto que vive en memoria, sin tocar el disco.*
- Llama a `df.to_csv(buffer, index=False)`.
- **Imprime** en pantalla: `buffer.getvalue()`.
- Llama a `buffer.seek(0)`.
- Asigna a `leido`: `pd.read_csv(buffer)`
- **Imprime** en pantalla: `leido`.
