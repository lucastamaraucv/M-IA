## Diccionarios avanzados

Union de diccionarios con `|`, comprehensions y `defaultdict`.

```python
base = {"a": 1, "b": 2}
extra = {"b": 20, "c": 3}
combinado = base | extra
print(combinado)

cuadrados = {n: n**2 for n in range(6) if n % 2 == 0}
print(cuadrados)

from collections import defaultdict
conteo = defaultdict(int)
for letra in "abracadabra":
    conteo[letra] += 1
print(dict(conteo))
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- **Imprime** en pantalla: `cuadrados`.
- Desde `collections` importa `defaultdict`.
- Asigna a `conteo`: `defaultdict(int)`
- Bucle **for**: recorre `'abracadabra'` guardando cada valor en `letra`:
    - Actualiza `conteo[letra]` (`conteo[letra] += 1`)

## functools: reduce, partial, lru_cache

Programacion funcional y memoizacion sin escribir cache a mano.

```python
from functools import reduce, partial, lru_cache

producto = reduce(lambda a, b: a * b, [1, 2, 3, 4])
print("Producto:", producto)

def potencia(base, exponente):
    return base ** exponente

cuadrado = partial(potencia, exponente=2)
print("Cuadrado de 9:", cuadrado(9))

@lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print("Fibonacci(20):", fib(20))
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Desde `functools` importa `reduce`, `partial`, `lru_cache`.
- Asigna a `producto`: `reduce(lambda a, b: a * b, [1, 2, 3, 4])` — *functools.reduce(fn, it) — acumula it aplicando fn de a pares hasta un solo resultado.*
- **Imprime** en pantalla: `'Producto:', producto`.
- Define la funcion `potencia(base, exponente)`:
    - **Devuelve** `base ** exponente`.
- Asigna a `cuadrado`: `partial(potencia, exponente=2)`
- **Imprime** en pantalla: `'Cuadrado de 9:', cuadrado(9)`.
- `@lru_cache(maxsize=None)` Define la funcion `fib(n)`:
    - **Si** `n < 2` entonces:
        - **Devuelve** `n`.
    - **Devuelve** `fib(n - 1) + fib(n - 2)`.
- **Imprime** en pantalla: `'Fibonacci(20):', fib(20)`.

## re: expresiones regulares basicas

Buscar patrones, capturar grupos y reemplazar texto.

```python
import re

texto = "Contacto: ana@mail.com, tel 555-1234, luis@empresa.org"

emails = re.findall(r"[\w.]+@[\w.]+", texto)
print("Emails:", emails)

m = re.search(r"tel (\d{3}-\d{4})", texto)
print("Telefono:", m.group(1) if m else None)

limpio = re.sub(r"\d", "#", texto)
print("Numeros ocultos:", limpio)
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Importa `re`.
- Asigna a `texto`: `'Contacto: ana@mail.com, tel 555-1234, luis@empresa.org'`
- Asigna a `emails`: `re.findall('[\\w.]+@[\\w.]+', texto)`
- **Imprime** en pantalla: `'Emails:', emails`.
- Asigna a `m`: `re.search('tel (\\d{3}-\\d{4})', texto)`
- **Imprime** en pantalla: `'Telefono:', m.group(1) if m else None`.
- Asigna a `limpio`: `re.sub('\\d', '#', texto)`
- **Imprime** en pantalla: `'Numeros ocultos:', limpio`.

## dataclasses: clases de datos sin boilerplate

`@dataclass` genera `__init__`, `__repr__` y `__eq__` automaticamente.

```python
from dataclasses import dataclass

@dataclass
class Punto:
    x: float
    y: float
    etiqueta: str = "sin nombre"

    def distancia_origen(self) -> float:
        return (self.x ** 2 + self.y ** 2) ** 0.5

p = Punto(3, 4)
print(p)
print(p.distancia_origen())
print(p == Punto(3, 4))
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Desde `dataclasses` importa `dataclass`.
- `@dataclass` Define la clase `Punto`:
    - Declara `x: float` = `(sin valor)`
    - Declara `y: float` = `(sin valor)`
    - Declara `etiqueta: str` = `'sin nombre'`
    - Define la funcion `distancia_origen(self)`:
        - **Devuelve** `(self.x ** 2 + self.y ** 2) ** 0.5`.
- Asigna a `p`: `Punto(3, 4)`
- **Imprime** en pantalla: `p`.
- **Imprime** en pantalla: `p.distancia_origen()`.
- **Imprime** en pantalla: `p == Punto(3, 4)`.

## context managers: with y contextlib

Como funciona `with` por dentro, usando `@contextmanager`.

```python
from contextlib import contextmanager
import time

@contextmanager
def cronometro(nombre):
    inicio = time.perf_counter()
    print(f"Iniciando {nombre}...")
    yield
    duracion = time.perf_counter() - inicio
    print(f"{nombre} tomo {duracion:.4f}s")

with cronometro("suma"):
    total = sum(range(1_000_000))

print("Total:", total)
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Desde `contextlib` importa `contextmanager`.
- Importa `time`.
- `@contextmanager` Define la funcion `cronometro(nombre)`:
    - Asigna a `inicio`: `time.perf_counter()`
    - **Imprime** en pantalla: `f'Iniciando {nombre}...'`.
    - **yield** `` — produce un valor y pausa la funcion hasta el proximo `next()`.
    - Asigna a `duracion`: `time.perf_counter() - inicio`
    - **Imprime** en pantalla: `f'{nombre} tomo {duracion:.4f}s'`.
- Bloque **with** usando `cronometro('suma')` (se cierra/libera automaticamente al salir):
    - Asigna a `total`: `sum(range(1000000))` — *range(a, b, paso) — secuencia perezosa de enteros, muy usada en bucles for.*
- **Imprime** en pantalla: `'Total:', total`.

## typing: anotaciones de tipos

Documentar (y dejar que herramientas como mypy verifiquen) los tipos esperados.

```python
from typing import Optional, Union

def buscar_usuario(id_usuario: int, base: dict[int, str]) -> Optional[str]:
    return base.get(id_usuario)

def normalizar(valor: Union[int, float, str]) -> float:
    return float(valor)

usuarios = {1: "Ana", 2: "Luis"}
print(buscar_usuario(1, usuarios))
print(buscar_usuario(99, usuarios))
print(normalizar("3.5"), normalizar(10))
```

**Paso a paso — que hace cada bloque de la celda de arriba:**

- Desde `typing` importa `Optional`, `Union`.
- Define la funcion `buscar_usuario(id_usuario: int, base: dict[int, str])`:
    - **Devuelve** `base.get(id_usuario)` — *dict.get(clave, default) — busca la clave sin lanzar KeyError si no existe.*.
- Define la funcion `normalizar(valor: Union[int, float, str])`:
    - **Devuelve** `float(valor)`.
- Asigna a `usuarios`: `{1: 'Ana', 2: 'Luis'}`
- **Imprime** en pantalla: `buscar_usuario(1, usuarios)`.
- **Imprime** en pantalla: `buscar_usuario(99, usuarios)`.
- **Imprime** en pantalla: `normalizar('3.5'), normalizar(10)`.
